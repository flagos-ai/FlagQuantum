# 02: Build a quantum training loop

You will define a one-qubit model, choose a target, inspect its gradient, and update its parameter yourself.
Then you will repeat that update with a short PyTorch loop.
No training helper is called: each part of the algorithm is in this notebook.

**Learning goal:** train a gate angle with a standard PyTorch gradient-and-optimizer loop.

**Prerequisites:** lesson 01 and basic PyTorch tensors, gradients and optimizers. All calculations run locally on CPU; no hardware account is required.

Choose a route in the [workshop guide](../README.md).


## 1. Define a circuit with an adjustable angle

The RY gate rotates a qubit by an angle `theta`. Our function describes how to build the circuit from the current parameter.
The optional `inputs` argument is unused in this small example; it can carry data in a larger model.

````python
%matplotlib inline
import matplotlib.pyplot as plt

import flagquantum as fq
import torch


def build_circuit(parameters, inputs=None):
    circuit = fq.Circuit(1)
    circuit.ry(0, parameters["theta"][0])
    return circuit
````

## 2. Make the angle trainable

`fq.Module` owns a trainable PyTorch parameter named `theta`. The shape `(1,)` means one number.
The seed gives us the same starting point when we restart this cell.

We ask the model to return the Z expectation of qubit 0. A Z measurement assigns +1 to `0` and -1 to `1`; its expectation is their average.
For this circuit, the answer is `cos(theta)`.

Each horizontal line in the circuit diagram is a qubit; gates execute from left to right. Vertical wire order is not the bit-string display order on a cloud platform. We draw only the initial-parameter circuit, not on every training iteration.


````python
policy = fq.RuntimePolicy(
    execution_options=fq.ExecutionOptions(device="cpu", mode="statevector"),
    observable="z_sum",
    observable_qubits=(0,),
)
model = fq.Module(
    build_circuit,
    parameters={"theta": (1,)},
    init="uniform",
    seed=42,
    policy=policy,
)
initial_circuit = build_circuit({name: value.detach() for name, value in model.parameter_groups.items()})
fig, ax = initial_circuit.draw(format="mpl", show_all_qubits=True)
plt.show()
for name, parameter in model.named_parameters():
    print(name, parameter.detach())
print("Initial prediction:", model().detach())
````

## 3. Choose what the model should learn

The target is 0.25. The loss is the squared difference between the prediction and the target.
**Before running:** what would a loss of zero tell you?

**Concept:** we train the RY gate angle, not the number of qubits. The calculation is `theta → cos(theta) → (cos(theta) - 0.25)²`.

**Expected output:** the prediction is a Z expectation in [-1, 1], and the loss should approach zero. For this one-qubit measurement, P(0)=(1+prediction)/2; the prediction itself is not a classification probability.

**Common misconception:** this is a local differentiable CPU simulation. `backward()` computes gradients locally; it does not submit a hardware task on every step.


````python
target = 0.25
prediction = model()
loss = (prediction - target).square().mean()
initial_loss = loss.item()
print("Prediction:", prediction.item())
print("Target:", target)
print("Loss:", initial_loss)

print("theta -> cos(theta) -> loss:", model.parameter_groups["theta"].detach().item(), prediction.item(), initial_loss)
````

## 4. Inspect the gradient

`backward()` computes how the loss changes with the angle.
The optimizer will use that gradient to adjust the parameter. Clearing old gradients matters because PyTorch accumulates them.

`zero_grad()` clears accumulated gradients before a new backward pass; otherwise previous steps contribute again. `detach()` disconnects the displayed or recorded tensor from its gradient graph, not the model itself. Do not detach `prediction` before calculating the training loss.


````python
learning_rate = 0.08
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
optimizer.zero_grad()
loss.backward()
for name, parameter in model.named_parameters():
    assert parameter.grad is not None
    assert torch.isfinite(parameter.grad).all()
    print(name, "gradient:", parameter.grad)
````

## 5. Take one update

This is the first optimizer step. Compare the new prediction with the old one.
To repeat the whole experiment, restart from the cell that creates `model`; rerunning `backward()` on an already-used graph is not a fresh training step.

````python
losses = [initial_loss]
optimizer.step()
print("Prediction after one update:", model().detach().item())
````

## 6. Repeat the same process

Each iteration makes a fresh prediction, computes its loss, differentiates it, and updates the angle.
We already took one step, so 79 more gives us 80 in total.

````python
total_steps = 80
for step in range(1, total_steps):
    optimizer.zero_grad()
    prediction = model()
    loss = (prediction - target).square().mean()
    loss.backward()
    optimizer.step()
    losses.append(loss.detach().item())

final_prediction = model().detach().item()
final_loss = (final_prediction - target) ** 2
print("Initial loss:", initial_loss)
print("Final loss:", final_loss)
print("Final prediction:", final_prediction)
assert final_loss < 0.01
````

## 7. See the learning process

The curve need not decrease on every step. With the default settings, the final loss should be below 0.01.

**Read the output:** `Final prediction` should approach 0.25 and `Final loss` should be below 0.01 with the supplied settings. Z=0.25 means P(0)=0.625 and P(1)=0.375 in this example, not a 25% class probability. A printed gradient is a sensitivity, not the next angle.

**Rerun safely:** recreate `model` and `optimizer` to restart training. Running the loop again continues updating the existing angle.


````python

plt.plot(losses)
plt.xlabel("Optimizer step")
plt.ylabel("Squared error")
plt.show()
````

## Take it home

The training pattern is ordinary PyTorch: build a circuit from a differentiable
parameter, calculate an expectation, backpropagate and update. This standalone
version makes the gradient path explicit.

```python
import torch
import flagquantum as fq

theta = torch.nn.Parameter(torch.tensor(0.7))
optimizer = torch.optim.Adam([theta], lr=0.08)
for step in range(80):
    optimizer.zero_grad()
    circuit = fq.Circuit(1).ry(0, theta)
    prediction = fq.run(circuit, outputs=fq.expectation(fq.Z(0))).expectation()
    loss = (prediction - 0.25).square().mean()
    loss.backward()
    optimizer.step()
print("Prediction:", prediction.detach().item())
```

Use `fq.Module` as in the lesson when you want a reusable PyTorch layer with
named parameters. A locally trained circuit is not a hardware training loop.
