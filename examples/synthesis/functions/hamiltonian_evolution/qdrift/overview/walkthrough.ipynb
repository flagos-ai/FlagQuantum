# Native qdrift workflow

Retain H = X(0)Y(1) + 0.5 Z(0), evolution coefficient 2.0 and all five samples. Native synthesis uses the fixed seed-zero schedule. Full-state acceptance uses an independent Circuit with the same sampled sequence; it does not assert equality to exact Hamiltonian evolution or prove sampling error bounds.

Run every cell from the repository root. See WORKFLOW.md for acceptance and scope.


```python
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import Output, QArray, QBit, allocate, qdrift, qfunc

hamiltonian = (fq.X(0) @ fq.Y(1)) + 0.5 * fq.Z(0)

@qfunc
def main(qba: Output[QArray[QBit]]) -> None:
    allocate(2, qba)
    qdrift(hamiltonian, 2.0, 5, qba)
```

```python
program = synth.synthesize(main)
result = fq.run(program)
probabilities = dict(program.probabilities(result, registers=('qba',)))
print(probabilities)
print(program.report["resources"])
```
