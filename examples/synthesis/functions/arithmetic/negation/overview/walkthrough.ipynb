# Exact signed negation

The original three-bit signed input spans -4 through 3 in superposition. The compiler infers an output wide enough for -(-4)=4; the user does not design a subtractor or decode two-complement bit strings.


```python
from collections import Counter
from random import Random
from typing import Annotated

import flagquantum as fq
from flagquantum.synth import Output, QNum, allocate, hadamard_transform, qfunc, synthesize
```

```python
@qfunc
def negate(a: Output[Annotated[QNum, (3, True, 0)]], b: Output[QNum]) -> None:
    allocate(a)
    hadamard_transform(a)
    b |= -a
```

```python
negate_program = synthesize(negate)
negate_result = fq.run(negate_program)
negate_probabilities = dict(negate_program.numeric_probabilities(negate_result, registers=('a', 'b')))
negate_counts = Counter(Random(17).choices(list(negate_probabilities), weights=list(negate_probabilities.values()), k=1000))
assert sum(negate_counts.values()) == 1000
print("Probabilities:", negate_probabilities)
print("Counts:", dict(negate_counts))
expected = {(a, -a): 1/8 for a in range(-4, 4)}
assert negate_probabilities.keys() == expected.keys()
assert all(abs(negate_probabilities[key]-value) < 1e-12 for key,value in expected.items())
```

## Interpretation

The two values are correlated quantum outputs. The probability table alone does not verify coherent phase; independent Circuit acceptance checks the complete complex state and clean workspace. Sampling is a local seeded replacement for the upstream cloud call.
