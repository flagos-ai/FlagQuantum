# Native mcx workflow

The retained original starts all seven controls at zero, so the target stays zero although the circuit contains the controlled operation. The supplemental coherent experiment below makes the operation observable and checks phase as well as probabilities. No claim of globally optimal MCX synthesis is made.

Run every cell from the repository root. See WORKFLOW.md for acceptance and scope.


```python
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import Output, QArray, QBit, X, allocate, control, qfunc

@qfunc
def main(cntrl: Output[QArray[QBit]], target: Output[QBit]) -> None:
    allocate(7, cntrl)
    allocate(target)
    control(cntrl, lambda: X(target))
```

```python
program = synth.synthesize(main)
result = fq.run(program)
probabilities = dict(program.probabilities(result, registers=('cntrl', 'target')))
print(probabilities)
print(program.report["resources"])
```

```python
from flagquantum.synth import H, RZ, S, hadamard_transform

@qfunc
def coherent(cntrl: Output[QArray[QBit]], target: Output[QBit]) -> None:
    allocate(7, cntrl)
    allocate(target)
    hadamard_transform(cntrl)
    S(cntrl[0])
    H(target)
    RZ(0.73, target)
    control(cntrl, lambda: X(target))

coherent_program = synth.synthesize(coherent)
coherent_result = fq.run(coherent_program)
print(dict(coherent_program.probabilities(coherent_result, registers=("cntrl", "target"))))
```
