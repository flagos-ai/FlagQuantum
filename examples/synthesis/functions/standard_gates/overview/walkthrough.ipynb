## Native imports for main

```python
import flagquantum as fq
from examples.synthesis.functions.standard_gates.crx.main import main as crx
from examples.synthesis.functions.standard_gates.cx.main import main as cx
from examples.synthesis.functions.standard_gates.phase.main import main as phase
from examples.synthesis.functions.standard_gates.r.main import main as r
from examples.synthesis.functions.standard_gates.rz.main import main as rz
from examples.synthesis.functions.standard_gates.rzz.main import main as rzz
from examples.synthesis.functions.standard_gates.swap.main import main as swap
from examples.synthesis.functions.standard_gates.u.main import main as u
from examples.synthesis.functions.standard_gates.x.main import main as x
from flagquantum import synth
```

## run example

```python
def run_example() -> None:
    """Reuse the native qfunc definitions rather than embed external source text."""
    for name, function in (
        ("X", x),
        ("RZ", rz),
        ("R", r),
        ("PHASE", phase),
        ("RZZ", rzz),
        ("CX", cx),
        ("CRX", crx),
        ("SWAP", swap),
        ("U", u),
    ):
        program = synth.synthesize(function)
        print("Gate:", name)
        print(program.report["resources"])
        result = fq.run(program)
        print(
            "Probabilities:",
            dict(program.probabilities(result, registers=tuple(program.registers))),
        )
```

## Execute the complete native workflow

Run every cell in order. Any sampling is local and seeded; resource reports come from the actual native program. Read the reported approximation limits and optimization outcomes alongside the numerical results.

```python
run_example()
```

## Verify absolute amplitudes and clean workspace

The independent Circuit uses the same declared register order. The comparison checks every complex amplitude without global-phase alignment and verifies zero probability in hidden work qubits. This is a local numerical check.


```python
import torch
from flagquantum import synth
from examples.synthesis.compare_states import compare_states
from examples.synthesis.functions.standard_gates.x.comparisons.circuit import (
    reference_circuit as x_reference,
)
from examples.synthesis.functions.standard_gates.rz.comparisons.circuit import (
    reference_circuit as rz_reference,
)
from examples.synthesis.functions.standard_gates.r.comparisons.circuit import (
    reference_circuit as r_reference,
)
from examples.synthesis.functions.standard_gates.phase.comparisons.circuit import (
    reference_circuit as phase_reference,
)
from examples.synthesis.functions.standard_gates.rzz.comparisons.circuit import (
    reference_circuit as rzz_reference,
)
from examples.synthesis.functions.standard_gates.cx.comparisons.circuit import (
    reference_circuit as cx_reference,
)
from examples.synthesis.functions.standard_gates.crx.comparisons.circuit import (
    reference_circuit as crx_reference,
)
from examples.synthesis.functions.standard_gates.swap.comparisons.circuit import (
    reference_circuit as swap_reference,
)
from examples.synthesis.functions.standard_gates.u.comparisons.circuit import (
    reference_circuit as u_reference,
)

for definition, reference_factory in (
    (x, x_reference),
    (rz, rz_reference),
    (r, r_reference),
    (phase, phase_reference),
    (rzz, rzz_reference),
    (cx, cx_reference),
    (crx, crx_reference),
    (swap, swap_reference),
    (u, u_reference),
):
    program = synth.synthesize(definition)
    compare_states(program, reference_factory(), tuple(program.registers))
print("Independent full-state comparison: passed")
```
