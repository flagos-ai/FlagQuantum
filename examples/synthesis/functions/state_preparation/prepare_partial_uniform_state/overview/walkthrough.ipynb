# Partial uniform state preparation

Retain the original four-bit interval [0,3) and five-bit interval [2,8), both sampling stages and both plots. Use native fresh-output preparation and 1000 seeded host samples from exact locally executed probabilities. This preserves fresh-input states, not the original unitary on arbitrary existing inputs. Plotting uses the optional Matplotlib dependency. Run all cells from the repository root. See WORKFLOW.md for acceptance and adaptations.


```python
from collections import Counter
from random import Random
import matplotlib.pyplot as plt

programs = {}
figures = {}

import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import (
    Output,
    QNum,
    allocate,
    prepare_uniform_interval_state,
    qfunc,
)


@qfunc
def trimmed(x: Output[QNum]) -> None:
    allocate(4, x)
    prepare_uniform_interval_state(0, 3, x)


programs["trimmed"] = synth.synthesize(trimmed)
```

```python
trimmed_program = programs["trimmed"]
trimmed_result = fq.run(trimmed_program)
trimmed_probabilities = dict(trimmed_program.probabilities(trimmed_result, registers=("x",)))
trimmed_counts = Counter(Random(17).choices(list(trimmed_probabilities), weights=list(trimmed_probabilities.values()), k=1000))
assert sum(trimmed_counts.values()) == 1000
print(dict(trimmed_counts))
```

```python
figure, axes = plt.subplots(figsize=(4, 3))
axes.bar([state[0] for state in sorted(trimmed_counts)], [count for state, count in sorted(trimmed_counts.items())], color="skyblue", edgecolor="black")
axes.set_xlabel("state")
axes.set_ylabel("shots")
figures["trimmed"] = figure
plt.show()
```

```python
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import (
    Output,
    QNum,
    allocate,
    prepare_uniform_interval_state,
    qfunc,
)


@qfunc
def interval(x: Output[QNum]) -> None:
    allocate(5, x)
    prepare_uniform_interval_state(2, 8, x)


programs["interval"] = synth.synthesize(interval)
```

```python
interval_program = programs["interval"]
interval_result = fq.run(interval_program)
interval_probabilities = dict(interval_program.probabilities(interval_result, registers=("x",)))
interval_counts = Counter(Random(17).choices(list(interval_probabilities), weights=list(interval_probabilities.values()), k=1000))
assert sum(interval_counts.values()) == 1000
print(dict(interval_counts))
```

```python
figure, axes = plt.subplots(figsize=(5, 3))
axes.bar([state[0] for state in sorted(interval_counts)], [count for state, count in sorted(interval_counts.items())], color="skyblue", edgecolor="black")
axes.set_xlabel("state")
axes.set_ylabel("shots")
figures["interval"] = figure
plt.show()
```
