# Qft in native FlagQuantum

Complete local computational Notebook workflow: allocate four qubits, apply QFT and synthesize. Local execution and a Fourier-matrix Circuit comparison additionally verify the program.

Use FlagQuantum qfunc and shared checked Fourier lowering. The source Notebook only constructs and synthesizes; local execution is an additional validation, with no cloud or training claim.

Define the native quantum functions below, synthesize them locally, inspect the executed results, and compare the complete complex state with an independent Circuit.

See [the native Python entry](main.py).

## Native imports for main

```python
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import Output, QArray, QBit, allocate, qft, qfunc
```

## main

```python
@qfunc
def main(x: Output[QArray[QBit]]) -> None:
    allocate(4, x)
    qft(x)
```

## build program

```python
def build_program() -> synth.Program:
    """Run the complete source Notebook's allocate/transform/synthesize workflow."""
    return synth.synthesize(main)
```

## Execute the complete native workflow

Run every cell in order. Any sampling is local and seeded; resource reports come from the actual native program. Read the reported approximation limits and optimization outcomes alongside the numerical results.

```python
program = build_program()
```

```python
result = fq.run(program)
```

```python
print(dict(program.probabilities(result, registers=("x",))))
```
