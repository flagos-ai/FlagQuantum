## Native stage 6

Counterpart of original code cell 14.


```python
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import (
    RX,
    Output,
    QArray,
    QBit,
    allocate,
    hadamard_transform,
    phase,
    qfunc,
)
from examples.synthesis.compare_states import compare_states
from examples.synthesis.tutorials.workshops.combinatorial_workshop.combinatorial_qmod_workshop_for_maxcut.comparisons.circuit import (
    reference_circuit,
)


@qfunc
def cut_program(params: list[float], v: Output[QArray[QBit]]) -> None:
    allocate(5, v)
    hadamard_transform(v)
    for i in range(4):
        phase(
            -(
                v[0] * (1 - v[1])
                + v[1] * (1 - v[0])
                + v[0] * (1 - v[2])
                + v[2] * (1 - v[0])
                + v[1] * (1 - v[2])
                + v[2] * (1 - v[1])
                + v[1] * (1 - v[3])
                + v[3] * (1 - v[1])
                + v[2] * (1 - v[4])
                + v[4] * (1 - v[2])
                + v[3] * (1 - v[4])
                + v[4] * (1 - v[3])
            ),
            params[2 * i],
        )
        mixer_layer(params[2 * i + 1], v)


preview = synth.synthesize(
    cut_program, classical_inputs={"params": initial_parameters()}, max_workspace=0
)
compare_states(preview, reference_circuit(initial_parameters()), ("v",))
print("Editable four-layer program:", preview.report["resources"])
layers = 4
print("Native complete quantum function:", main)
print("Layers:", layers, "physical bits:", 5)
```
