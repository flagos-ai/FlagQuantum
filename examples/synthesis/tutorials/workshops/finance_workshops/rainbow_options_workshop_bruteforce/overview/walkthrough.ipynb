## Original inputs


```python
from IPython.display import Image, display

from examples.synthesis.applications.finance.rainbow_options.rainbow_options_bruteforce_method.overview.model import (
    ALPHA,
    ASSERTION_ALPHA,
    EPSILON,
    SOURCE_PRICE,
    STRIKE,
    fresh_model,
)
from examples.synthesis.tutorials.workshops.finance_workshops.rainbow_options_workshop_bruteforce.overview.main import (
    report,
    save_figure,
)
from examples.synthesis.tutorials.workshops.finance_workshops.rainbow_options_workshop_bruteforce.overview.workflow import (
    SHOTS,
    build_program,
    maximum_asset_price,
    run_pricing,
)
```

## Build the correlated model


```python
model = fresh_model()
print("Rounded Cholesky rows:", model.coefficients)
```

## Four original Gaussian bins


```python
print("Grid and bin masses:", model.grid, model.probabilities)
```

## Strike guard


```python
assert maximum_asset_price(model) >= STRIKE
print("Largest reachable asset price:", maximum_asset_price(model))
```

## One fractional bit


```python
fraction_bits = 1
print("Affine precision:", 1 / (1 << fraction_bits))
```

## Generate the affine rows and relative offset


```python
print("Affine rows and relative offset:", model.coefficients, model.offset)
```

## Inspect the maximum selector


```python
print("Joint selectors:", model.selectors)
```

## Build the normalized payoff function


```python
print("Raw grid and payoff scale:", model.payoff_grid, model.scale)
```

## Load square-root payoff amplitudes


```python
print("Joint indicator rotations:", model.angles)
```

## The two asset words


```python
assert len(model.probabilities) == 4 and len(model.selectors) == 16
```

## Compose the state preparation


```python
preparation = build_program(model, max_width=14)
```

## Inspect the named output


```python
print("Outputs:", preparation.registers)
```

## Check the preparation width


```python
print("Preparation resources:", preparation.report["resources"])
```

## Count-only estimation settings


```python
print("Precision and confidence:", EPSILON, 1 - ALPHA)
```

## Bind the estimation width


```python
estimation = build_program(model, max_width=25)
```

## Inspect estimation resources


```python
print("Estimation resources:", estimation.report["resources"])
```

## Take the original 20,000 shots


```python
result = run_pricing()
report(result)
assert SHOTS == 20_000
```

## Three-bit selector grid


```python
assert len(model.payoff_grid) == 8 and max(model.selectors) < 8
```

## Rescale the interval


```python
print("Observed price interval:", result.price_interval)
```

## Compare against the exact finite model


```python
display(Image(filename=str(save_figure(result))))
```

## Check the historical reference


```python
print("Reference confidence level:", 1 - ASSERTION_ALPHA)
print(
    "Within source tolerance:",
    abs(result.price - SOURCE_PRICE) <= result.source_tolerance,
)
```
