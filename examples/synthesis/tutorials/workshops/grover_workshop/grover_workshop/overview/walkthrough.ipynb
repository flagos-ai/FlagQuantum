## Complete the search instead of mistaking allocation for it

The two drafts intentionally allocate seven qubits without an oracle or
reflection. The editable solution below composes the arithmetic marking
predicate and the diffusion step. FlagQuantum supplies reversible comparisons
and coherent phase construction; the user chooses the equation and one Grover
round. Both width/depth objective routes are retained for draft and solution.

Try a different equation only with an updated independent reference. Compare
solution probability across all satisfying assignments, not only the most
frequent output. More Grover rounds are not monotonically better. Allocation
is a useful starting exercise but is explicitly not counted as a search.


```python
"""Retain the Grover workshop's uniform state, two drafts and two solutions."""

from typing import Annotated

import flagquantum as fq
from examples.synthesis.tutorials.workshops.grover_workshop.grover_workshop.main import (
    main as solution,
)
from flagquantum import synth
from flagquantum.synth import (
    Output,
    QArray,
    QBit,
    QNum,
    allocate,
    hadamard_transform,
    qfunc,
)


@qfunc
def uniform(x: Output[QArray[QBit]]) -> None:
    allocate(5, x)
    hadamard_transform(x)


@qfunc
def draft(
    a: Output[Annotated[QNum, 2]],
    b: Output[Annotated[QNum, 2]],
    c: Output[Annotated[QNum, 3]],
) -> None:
    allocate(a)
    allocate(b)
    allocate(c)


def run_workflow() -> dict[str, synth.Program]:
    programs = {"uniform": synth.synthesize(uniform, max_workspace=0)}
    for name, callback in (("draft", draft), ("solution", solution)):
        for objective in ("width", "depth"):
            programs[f"{name}_{objective}"] = synth.synthesize(
                callback, objective=objective, max_workspace=0
            )
    for name, program in programs.items():
        registers = ("x",) if name == "uniform" else ("a", "b", "c")
        print(name, dict(program.probabilities(fq.run(program), registers=registers)))
        print("Resources:", program.report["resources"])
    return programs


def run_example() -> None:
    run_workflow()


from math import pi
from flagquantum.synth import control, phase
from examples.synthesis.tutorials.workshops.grover_workshop.grover_workshop.main import (
    mark,
    diffuse,
)


@qfunc
def solution(
    a: Output[Annotated[QNum, 2]],
    b: Output[Annotated[QNum, 2]],
    c: Output[Annotated[QNum, 3]],
) -> None:
    allocate(a)
    allocate(b)
    allocate(c)
    hadamard_transform(a)
    hadamard_transform(b)
    hadamard_transform(c)
    mark(a, b, c)
    diffuse(a, b, c)
```

```python
programs = run_workflow()
```

```python
print("Exercise drafts deliberately contain allocation only.")
print("Solution builds:", [name for name in programs if name.startswith("solution")])
```
