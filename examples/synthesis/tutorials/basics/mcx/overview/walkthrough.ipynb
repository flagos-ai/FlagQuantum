## Express the condition, constrain the implementation

The editable quantum function below uses a whole logical control array. It
prepares an optional basis witness and expresses an all-ones controlled flip.
Static `size` is allocation size; `enabled` chooses how many inputs start at one.
The compiler builds the MCX and cleans work storage within the declared budgets.

The original zero-input run is intentionally inactive. Later checks include
active and coherent witnesses, so unchanged zero output is not treated as proof
of controlled action. Try binding `enabled=14` in the fourteen-control program:
the target should flip. Resource comparisons and the 50-control acceptance
remain the original problems. The Boston snapshot is a local routing target,
not a live hardware submission or a globally optimal compilation claim.


Import the native tutorial and its source qfunc.

```python
import examples.synthesis.tutorials.basics.mcx.overview.main as report
import examples.synthesis.tutorials.basics.mcx.overview.workflow as workflow
from examples.synthesis.tutorials.basics.mcx.main import my_mcx
from flagquantum.synth import synthesize
from flagquantum.synth import Output, QArray, QBit, X, allocate, control, qfunc

@qfunc
def conditional_flip(size: int, enabled: int, initial_target: int,
                     controls: Output[QArray[QBit]], target: Output[QBit]) -> None:
    allocate(size, controls)
    allocate(target)
    for bit in range(enabled):
        X(controls[bit])
    if initial_target == 1:
        X(target)
    control(controls, lambda: X(target))
```

Use the same quantum control function at both source sizes.

```python
kernel = my_mcx
```

Synthesize under the original 20-qubit depth budget.

```python
depth_program = synthesize(conditional_flip, classical_inputs={"size": controls, "enabled": 0, "initial_target": 0}, objective="depth", max_qubits=20, max_workspace=5, memory_limit_bytes=128 * 1024**2)
```
