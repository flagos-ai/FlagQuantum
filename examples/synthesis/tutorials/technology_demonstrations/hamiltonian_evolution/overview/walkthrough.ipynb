# Water Hamiltonian evolution benchmark

Compare seven original-sized schedules against independently executed full-state references.

Regenerate the original water Hamiltonian from its STO-3G model.

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.tutorials.technology_demonstrations.hamiltonian_evolution.overview import benchmark
from examples.synthesis.tutorials.technology_demonstrations.hamiltonian_evolution.overview.main import print_model, print_results, save_figure

model = benchmark.fresh_model()
print_model(model)
```

Keep orders 1, 2, and 4 with 6, 4 and 1 repeated steps, and both qDRIFT sizes.

```python
print('Product-formula settings:', ((1, 6), (2, 4), (4, 1)))
print('qDRIFT draws:', (1000, 2000))
```

Prepare the seven-row resource report.

```python
print('Expected benchmark rows:', 7)
```

Use the original named CX/U basis for all current metrics.

```python
print('Named target basis:', benchmark.TARGET_BASIS)
```

Execute all three product formulas and check independent complete states.

```python
trotter, controlled = benchmark.run_trotter(model)
print('Product formulas:', tuple(row.label for row in trotter))
```

Inspect both controlled products, including the coherent superposition of branches.

```python
print('Controlled products:', tuple(row.label for row in controlled))
assert all(row.target.n_qubits == 13 for row in controlled)
```

Synthesize, sample schedules and verify qDRIFT with 1,000 and 2,000 terms.

```python
sampled = benchmark.run_qdrift(model)
assert len(sampled) == 2
print('Sampled schedules:', tuple(row.label for row in sampled))
```

Load the seven historical Qiskit resource rows recorded by the upstream Notebook.

```python
print('Source-reported CX:', benchmark.HISTORICAL_QISKIT_CX)
print('Source-reported depths:', benchmark.HISTORICAL_QISKIT_DEPTH)
```

The source's Qiskit setup was provided as a commented reproduction note.

```python
print('Historical comparison basis:', benchmark.TARGET_BASIS)
```

Use the same fresh Hamiltonian for product and sampled evolution.

```python
assert len(model.terms) == 551
```

Check the six-step first-order product against its independent factor sequence.

```python
assert trotter[0].reference_error < 2e-10
```

Check the second-order and five-part fourth-order symmetric products.

```python
assert max(row.reference_error for row in trotter[1:]) < 2e-10
```

Check the two controlled native products after common-basis conversion.

```python
assert max(row.conversion_error for row in controlled) < 2e-10
```

Check both sampled sequences against independent inverse-CDF reference circuits.

```python
assert max(row.reference_error for row in sampled) < 2e-10
```

Report actual local counts and the source-reported comparator separately.

```python
rows = (*trotter, *controlled, *sampled)
print_results(rows)
```

Plot the CX/U counts and depths for all seven cases.

```python
figures = Path('notebook-figures')
save_figure(rows, figures)
display(Image(filename=str(figures / 'evolution-cx-depth.png')))
```
