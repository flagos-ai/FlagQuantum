# Water Hamiltonian evolution benchmark

Compare seven original-sized schedules against independently executed full-state references.

## Identify what the evolution request compiles

The native builders in [benchmark.py](benchmark.py) bind a fresh 551-term,
twelve-qubit water Hamiltonian into qfunc evolution requests. FlagQuantum
constructs the elementary Pauli rotations; the host then composes repeated
steps, higher-order products and controlled Circuit variants. The complete
seven-row experiment is a mixed synth/Circuit benchmark, not one whole-program
high-level synthesis call.

The source schedules, both 1000/2000-term qDRIFT sequences and their independent
full-state checks remain. Changing order or sample count requires the matching
reference sequence; qDRIFT circuit agreement checks identical random choices,
not equality to exact continuous evolution. Current resources are converted to
the named CX/U basis. Historical Qiskit rows remain source-reported values,
not a fresh same-machine benchmark or measured hardware timing. Try comparing
error and converted cost across the existing rows before altering a schedule.
