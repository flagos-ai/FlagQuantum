## Distinguish a high-level root from a search harness

The visible root below requests one coin-controlled modular shift using shared
[reversible helpers](quantum.py). The benchmark compares this qfunc family at
several workspace caps with a separately coded Circuit parity construction.
The host enumerates candidates and selects their measured CX count; automatic
global optimization by `synthesize` is not claimed.

All seven original widths, endpoint state checks and CX/U3 conversions remain.
Try another admitted preview width and inspect both physical width and actual
CX count. The local timing curves measure current construction and bounded
search. The thirteen published large-instance rows through sixty position bits
are preserved historical data, not newly executed local circuits.


Import the native benchmark and plotting stages.

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.tutorials.technology_demonstrations.discrete_quantum_walk_circle.overview import (
    archive,
    benchmark,
    plots,
)
from examples.synthesis.tutorials.technology_demonstrations.discrete_quantum_walk_circle.overview.main import (
    report,
    verify,
)

figures = Path("notebook-figures")
from flagquantum import synth
from flagquantum.synth import H, Output, QBit, QNum, allocate, control, qfunc
from examples.synthesis.tutorials.technology_demonstrations.discrete_quantum_walk_circle.overview.quantum import (
    increment,
    decrement,
)


@qfunc
def one_step(width: int, position: Output[QNum], coin: Output[QBit]) -> None:
    allocate(width, position)
    allocate(coin)
    H(coin)
    control(coin == 0, lambda: increment(width, position))
    control(coin == 1, lambda: decrement(width, position))


preview = synth.synthesize(
    one_step,
    classical_inputs={"width": 5},
    max_qubits=10,
    max_workspace=4,
    objective="width",
    max_operations=250000,
    memory_limit_bytes=128 * 1024**2,
)
print("Preview logical shift resources:", preview.report["resources"])
```
