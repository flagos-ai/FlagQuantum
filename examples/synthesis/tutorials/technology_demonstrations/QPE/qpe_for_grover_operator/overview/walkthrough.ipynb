# Grover operator QPE: complete original precision study

Retain four data bits, one indicator, the six original marked integers, zero
initial data, all phase precisions one through five, both width/depth objectives
and the original depth-mode bounds 15+3(precision-1). No amplitude-preparation call
is added before QPE: the source intentionally starts its private data at zero.

The complete Grover unitary includes the source constant global minus sign.
Its controlled interference matters. Independent full physical Circuit checks
retain the dropped private data as an entangled environment. It is not called
clean compiler workspace. Target costs are actual exact CX/U3 measurements;
Qiskit arrays are source historical data. Current native width/depth objectives
produce equal resources here, which is reported rather than presented as an
optimization gain. The default source synthesis is represented by the width
objective: each native policy still executes and is measured separately.


```python
import math
from pathlib import Path
from typing import Annotated, Literal

from examples.synthesis.tutorials.technology_demonstrations.QPE.qpe_for_grover_operator.overview.ansatz import (
    build_program,
)
from examples.synthesis.tutorials.technology_demonstrations.QPE.qpe_for_grover_operator.overview.workflow import (
    plot_study,
    verify_program,
)
from IPython.display import Image, display

from flagquantum import synth
from flagquantum.synth import (
    H,
    Output,
    QBit,
    QNum,
    X,
    adjoint,
    allocate,
    apply,
    control,
    drop,
    hadamard_transform,
    phase,
    qfunc,
    qpe,
    within,
)

output_directory = Path("artifacts/development/grover_qpe")
```

```python
marked_integers = (2, 3, 5, 7, 11, 13)
precisions = tuple(range(1, 6))
print("Original data bits: 4; indicator bits: 1")
print("Original marked integers:", marked_integers)
print("Original precisions:", precisions)
```

```python
print("Native target basis: exact CX/U3")
print("Source Qiskit optimization level: 3 (historical, not rerun)")
```

```python
@qfunc
def state_preparation(x: QNum, indicator: QBit) -> None:
    hadamard_transform(x)
    with within():
        H(indicator)
    with apply():
        control(
            ((x >= 2) & (x <= 2))
            | ((x >= 3) & (x <= 3))
            | ((x >= 5) & (x <= 5))
            | ((x >= 7) & (x <= 7))
            | ((x >= 11) & (x <= 11))
            | ((x >= 13) & (x <= 13)),
            lambda: phase(indicator, math.pi),
        )
```

```python
@qfunc
def undo_preparation(x: QNum, indicator: QBit) -> None:
    with adjoint():
        state_preparation(x, indicator)


@qfunc
def grover(x: QNum, indicator: QBit) -> None:
    phase(indicator, math.pi)
    with within():
        undo_preparation(x, indicator)
    with apply():
        with within():
            X(indicator)
        with apply():
            control(x <= 0, lambda: phase(indicator, math.pi))
    phase(1.0, math.pi)
```

```python
print("All actual native resources use operator-exact public target conversion")
print(
    "No source SaaS preferences or timeout settings are claimed as native optimizer behavior"
)
```

```python
@qfunc
def precision_one(phase_result: Output[Annotated[QNum, (1, False, 1)]]) -> None:
    indicator: QBit = QBit()
    x: Annotated[QNum, 4] = QNum()
    allocate(indicator)
    allocate(x)
    allocate(phase_result)
    qpe(lambda: grover(x, indicator), phase_result)
    drop(x)
    drop(indicator)


visible_program = synth.synthesize(precision_one, max_qubits=15, max_workspace=0)
print("Visible user root:", verify_program(visible_program, 1, "width"))
objectives: tuple[Literal["width", "depth"], ...] = ("width", "depth")
points = []
for precision in precisions:
    for objective in objectives:
        program = build_program(precision, objective)
        point = verify_program(program, precision, objective)
        points.append(point)
        print(point)
print(
    "Depth-mode source bounds:", [15 + 3 * (precision - 1) for precision in precisions]
)
```
