# Fixed-point amplitude amplification with fresh QSVT phases

The source arithmetic problem has a two-bit field a, a one-bit field b and a three-bit field c. Start in a uniform superposition and mark the words satisfying a+b=3 and c−a=2. The native oracle enumerates the complete 64-word classical truth table at synthesis time and specializes its two disjoint matches. The quantum program applies all 26 alternating projected rotations and uncomputes its predicate workspace.

Fit an odd degree-25 Chebyshev polynomial to the source 0.95 sign target on overlaps at least 0.1. The local implementation uses a Remez exchange on a 20,001-point grid, then fits symmetric projected rotations using a seeded numerical warm-up and coefficient collocation. It verifies the fresh phase response against the independently fitted polynomial. The degree and domain are those of the original example. The measured approximation error comes from the local fit.

Run the native program and draw 2,048 local samples. Compare the actual accepted count with the binomial expectation derived from the polynomial and check the original five-standard-deviation rule. Run with FlagQuantum, NumPy, Matplotlib and Jupyter installed.


## Local setup

```python
from importlib import import_module
```

```python
from examples.synthesis.algorithms.amplitude_amplification.qsvt_fixed_point.overview.fitting import (
    MINIMUM,
    TARGET,
    fit_source,
)
from examples.synthesis.algorithms.amplitude_amplification.qsvt_fixed_point.overview.main import (
    OUTPUT,
)
from examples.synthesis.algorithms.amplitude_amplification.qsvt_fixed_point.overview.plots import (
    save_polynomial,
)
from examples.synthesis.algorithms.amplitude_amplification.qsvt_fixed_point.overview.quantum import (
    build_program,
    matching_words,
)
from examples.synthesis.algorithms.amplitude_amplification.qsvt_fixed_point.overview.workflow import (
    SHOTS,
    run_search,
    show_fit,
    show_outcome,
)
```

## Starting-state projector

```python
field_widths = (2, 1, 3)
state_count = 2 ** sum(field_widths)
print("Uniform input support:", state_count)
```

## Arithmetic target projector

```python
targets = matching_words()
print("Arithmetic matches:", targets)
```

```python
assert all(a + b == 3 and c - a == 2 for a, b, c in targets)
print("Original match count:", len(targets))
```

## Alternating projected phases

```python
input_overlap = (len(targets) / state_count) ** 0.5
assert input_overlap >= MINIMUM
print("Initial overlap:", input_overlap)
```

## Fit the original sign polynomial and physical phases

```python
fit = fit_source()
show_fit(fit)
plot = save_polynomial(fit, OUTPUT)
viewer = import_module("IPython.display")
viewer.display(viewer.Image(filename=str(plot)))
```

```python
print("Target scale:", TARGET)
print("Fresh phase sequence:", fit.phases)
```

```python
program = build_program(fit.phases)
print("Named output ports:", tuple(program.registers))
print("Synthesized resources:", program.report["resources"])
```

## Measure and compare against the fitted polynomial

```python
print("Original sample budget:", SHOTS)
outcome = run_search(fit)
```

```python
show_outcome(outcome)
```

```python
assert abs(outcome.actual_probability - outcome.predicted_probability) < 1e-10
```

```python
assert abs(outcome.good - outcome.expected) <= 5 * outcome.sigma
```
