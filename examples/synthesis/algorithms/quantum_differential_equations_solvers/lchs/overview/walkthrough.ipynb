# Linear combination of Hamiltonian simulations

Recompute the source 64-point signed kernel grid and run its two comparisons. The classical experiment uses the full 128-dimensional random PSD and Hermitian matrices and evolves to time 10. Two fixed local seeds replace the original unseeded matrix draws. The quantum experiment uses the original two data qubits, six grid bits, three block flags and one signal bit.

The native program prepares the kernel, applies the original signed linear block and the Pauli SELECT paths, performs fresh GQSP rotations, applies the kernel phase and uncomputes kernel preparation. The phase degree is selected from the source Bessel-tail rule. The native program uses synthesis for amplitude preparation, signed phase and the controlled rotation table, and the public Circuit API for coherent GQSP assembly.

Compare the zero-flag projection with all 64 independently exponentiated grid terms and with the dense nonunitary reference. Report the original fidelity and amplitude norm and the native absolute projection error. Run with FlagQuantum, NumPy, Matplotlib and Jupyter installed.
