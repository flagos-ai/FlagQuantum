# Qubitized Hamiltonian evolution

All 15 source stages, the original four Pauli terms on two data qubits, time 22, degree 33 and 10 qubits for evolution. Compare both complex projected vectors using the declared scales.

Recompute the Jacobi-Anger coefficients by cosine quadrature at the source degree. Native synth prepares the amplitudes at double precision. Public Circuit SELECT, reflection and inverse preparation implement the 63 controlled walk steps. Use a reproducible local input seed in place of the unseeded source draw. Circuit resources describe the emitted native IR.

## Load the native verification workflow

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.algorithms.hamiltonian_simulation.hamiltonian_simulation_with_block_encoding.hamiltonian_simulation_qubitization.overview.model import (
    ALPHA,
    DEGREE,
    TERMS,
    TIME,
    TOLERANCE,
    expansion,
    fresh_input,
)
from examples.synthesis.algorithms.hamiltonian_simulation.hamiltonian_simulation_with_block_encoding.hamiltonian_simulation_qubitization.overview.plots import (
    save_comparison,
)
from examples.synthesis.algorithms.hamiltonian_simulation.hamiltonian_simulation_with_block_encoding.hamiltonian_simulation_qubitization.overview.quantum import (
    EVOLUTION_FLAGS,
    HAMILTONIAN_FLAGS,
    controlled_walk,
)
from examples.synthesis.algorithms.hamiltonian_simulation.hamiltonian_simulation_with_block_encoding.hamiltonian_simulation_qubitization.overview.workflow import (
    run_workflow,
)

folder = Path("artifacts/development/qubitized-hamiltonian")
```

## Keep the original Pauli weights and time

```python
print("Terms:", TERMS)
print("Time and Chebyshev degree:", TIME, DEGREE)
```

## Read the Hamiltonian flag width and scale

```python
print("Hamiltonian flags and scale:", HAMILTONIAN_FLAGS, ALPHA)
```

## Draw a fresh normalized source-style input

```python
print("Fresh input:", fresh_input().tolist())
```

## Inspect the target evolution tolerance

```python
print("Evolution tolerance:", TOLERANCE)
```

## Regenerate the complex polynomial

```python
poly = expansion()
assert len(poly.coefficients) == DEGREE + 1
```

## Read the evolution flag width and scale

```python
print("Evolution flags:", EVOLUTION_FLAGS)
print("Scales:", ALPHA, poly.scale)
```

## Construct the native controlled walk

```python
walk = controlled_walk()
assert walk.n_qubits == 5
```

## Execute both checked projections

```python
hamiltonian, evolution = run_workflow()
```

## Compare the Hamiltonian projection

```python
print("pauli error:", hamiltonian.error)
print("pauli width:", hamiltonian.circuit.n_qubits)
print("pauli IR instructions:", len(hamiltonian.circuit.to_ir().instructions))
print("pauli scaled amplitudes:", (hamiltonian.raw * hamiltonian.scale).tolist())
display(
    Image(
        filename=str(
            save_comparison(hamiltonian, "Pauli block encoding", folder / "pauli.png")
        )
    )
)
```

## Select the controlled powers

```python
print("Walk powers:", [2**i for i in range(EVOLUTION_FLAGS)])
```

## Inspect the complex coefficient phases

```python
print("First four coefficients:", poly.coefficients[:4])
```

## Inspect the actual emitted evolution Circuit

```python
print("evolution width:", evolution.circuit.n_qubits)
print("evolution IR instructions:", len(evolution.circuit.to_ir().instructions))
```

## Compare the absolute evolution amplitudes

```python
print("evolution scaled amplitudes:", (evolution.raw * evolution.scale).tolist())
```

## Verify the time evolution

```python
print("evolution error:", evolution.error)
print("Evolution fidelity:", evolution.fidelity)
display(
    Image(
        filename=str(
            save_comparison(
                evolution, "Qubitized time evolution", folder / "evolution.png"
            )
        )
    )
)
```
