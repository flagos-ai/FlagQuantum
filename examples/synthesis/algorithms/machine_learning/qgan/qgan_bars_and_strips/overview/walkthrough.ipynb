## Distinguish fixed qfunc construction from differentiable training

The visible qfunc preview prepares the same four-qubit encoding and eight-weight
ansatz used by the training layer. It exposes static specialization and named
readout. The full minibatch training instead uses the PyTorch-tensor Circuit
path in [neural.py](neural.py), retaining batched differentiable parameters.
Replacing that path with Python floats in a synthesizer would lose its gradient
contract. The two paths are explicitly compared on complete complex states.

The source rounds generator outputs. Its zero generator gradients and unchanged
weights are reported, while discriminator weights do update. Do not interpret
perfect bars-and-stripes membership as learned diversity or claim a trained
quantum generator when only the discriminator changed. The local finite-shot
forward value uses an exact-observable derivative before rounding, not an
unbiased sample-based gradient estimator.

Try a nonzero feature/weight binding in the preview and compare it with the
batched layer. Keep the original ten classical epochs, one hybrid epoch and
checkpoint hashes for acceptance. Removing round to enable generator learning
would be a separate experiment, requiring new evidence rather than rewriting
the source reproduction.

Use the native differentiable four-qubit circuit.

```python
import math
import torch
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import RX, RZ, RZZ, Output, QArray, QBit, R, allocate, qfunc


@qfunc
def encode(data: list[float], q: QArray[QBit]) -> None:
    for index in range(len(data)):
        RX(math.pi * data[index], q[index])
    for index in range(len(data)):
        RZ(math.pi * data[index], q[index])


@qfunc
def ansatz(weights: list[float], q: QArray[QBit]) -> None:
    for index in range(len(q)):
        R(weights[index], math.pi / 2, q[index])
    for index in range(len(q) - 1):
        RZZ(weights[len(q) + index], q[index : index + 2])
    RZZ(weights[7], q[0:2])


@qfunc
def generator_state(
    input_vec: list[float], weight: list[float], result: Output[QArray[QBit]]
) -> None:
    allocate(4, result)
    encode(input_vec, result)
    ansatz(weight, result)


preview_data = [0.17, 0.38, 0.53, 0.79]
preview_weights = [0.1 * (index + 1) for index in range(8)]
preview = synth.synthesize(
    generator_state,
    classical_inputs={"input_vec": preview_data, "weight": preview_weights},
    max_qubits=4,
    max_workspace=0,
    memory_limit_bytes=1024**2,
)
preview_result = fq.run(preview)
print(
    "Native qfunc distribution:",
    dict(preview.probabilities(preview_result, registers=("result",))),
)
batch_circuit = neural.batched_state
torch.testing.assert_close(
    preview_result.statevector().flatten(),
    batch_circuit(
        torch.tensor([preview_data], dtype=torch.float64),
        torch.tensor(preview_weights, dtype=torch.float64),
    ).flatten(),
    atol=3e-12,
    rtol=0,
)
```
