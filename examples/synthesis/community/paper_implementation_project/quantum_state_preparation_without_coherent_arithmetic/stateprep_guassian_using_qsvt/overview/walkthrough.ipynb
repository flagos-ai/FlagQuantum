# Gaussian preparation with QSVT

Regenerate both parity responses and measure the success amplitude of the eight-bit program before exact amplification.

Use the local FlagQuantum environment.

```python
from pathlib import Path

from IPython.display import Image, display

import flagquantum as fq
```

Import the native QSVT and amplification workflow.

```python
from examples.synthesis.community.paper_implementation_project.quantum_state_preparation_without_coherent_arithmetic.stateprep_guassian_using_qsvt.overview import (
    workflow,
)
from examples.synthesis.community.paper_implementation_project.quantum_state_preparation_without_coherent_arithmetic.stateprep_guassian_using_qsvt.overview.main import (
    print_report,
    save_figures,
)

print(fq.__name__)
```

Keep eight input bits and 256 coordinates on the Gaussian interval.

```python
print("Input bits and points:", workflow.WIDTH, workflow.SIZE)
```

Apply the original controlled sine rotations.

```python
print("Controlled angles:", workflow.ANGLES)
```

Use the Gaussian evaluated after the inverse sine coordinate map.

```python
print("Gaussian interval:", (-2, 2))
```

Preserve both parity components and their original degrees.

```python
print("Even and odd degrees:", workflow.EVEN_DEGREE, workflow.ODD_DEGREE)
```

Regenerate coefficients and the complete symmetric phase sequences.

```python
result = workflow.run()
print("Even and odd phase fit errors:", result.even.residual, result.odd.residual)
```

Check the native preparation against the independent signal-block reference.

```python
assert result.prep_error < 1e-10
```

Compare the continuum success amplitude with the discrete circuit.

```python
print("Continuum and measured success amplitudes:", result.continuum_overlap, result.measured_overlap)
```

Calibrate both exact amplification rounds from the measured amplitude.

```python
assert abs(result.accepted_mass - 1.0) < 1e-10
```

Verify the complete state and the restored compiler workspace.

```python
assert result.amplified_error < 1e-10
assert result.work_probability < 1e-22
```

Report actual native and CX/U3 counts and inspect the 256 output probabilities.

```python
print_report(result)
figures = Path("notebook-figures")
save_figures(result, figures)
display(Image(filename=str(figures / "gaussian-probability.png")))
display(Image(filename=str(figures / "gaussian-residual.png")))
```
