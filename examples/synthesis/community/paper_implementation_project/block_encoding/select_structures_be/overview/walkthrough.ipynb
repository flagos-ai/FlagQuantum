# Structured matrix block encodings

All ten source code stages and the original checkerboard, Toeplitz, symmetric tridiagonal, and 4 by 4 grid Laplacian models. Preserve the dimensions 8, 16, 16, and 16, the respective scales 8, 4, 3, and 5, and the original uniform input.

Generate signed coefficients from a reproducible local seed. Use native synth for amplitudes and a bounded public Circuit SELECT for each structured address map. Compare absolute complex amplitudes directly to each declared matrix product. The native Laplacian encodes the same matrix scaled by the magnitude of its diagonal. The historical checkerboard and Laplacian prints show opposite overall signs between their projected and theoretical vectors. This workflow preserves the signed theoretical matrices and verifies their native projections directly.

## Load the native structured encodings

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.community.paper_implementation_project.block_encoding.select_structures_be.overview.model import (
    make_structures,
)
from examples.synthesis.community.paper_implementation_project.block_encoding.select_structures_be.overview.plots import (
    save_projections,
)
from examples.synthesis.community.paper_implementation_project.block_encoding.select_structures_be.overview.workflow import (
    run_workflow,
)

output = Path("artifacts/development/structured-blocks/projections.png")
models = make_structures()
```

## Regenerate the signed checkerboard

```python
print(models[0].matrix.real.tolist())
```

## Inspect checkerboard addresses

```python
print(
    "Checkerboard size and selectors:",
    models[0].matrix.shape[0],
    models[0].selector_bits,
)
```

## Run and verify all four unnormalized projections

```python
results = run_workflow()
row = results[0]
print(
    "checkerboard shape, scale and width:",
    tuple(row.structure.matrix.shape),
    row.scale,
    row.circuit.n_qubits,
)
print("checkerboard unnormalized projection:", row.actual.tolist())
print("checkerboard absolute error:", row.error)
```

## Regenerate the original Toeplitz diagonals

```python
print(models[1].matrix.real.tolist())
```

## Verify the flagged Toeplitz boundary

```python
row = results[1]
print(
    "toeplitz shape, scale and width:",
    tuple(row.structure.matrix.shape),
    row.scale,
    row.circuit.n_qubits,
)
print("toeplitz unnormalized projection:", row.actual.tolist())
print("toeplitz absolute error:", row.error)
```

## Regenerate the signed symmetric tridiagonal

```python
print(models[2].matrix.real.tolist())
```

## Verify the three-address tridiagonal block

```python
row = results[2]
print(
    "tridiagonal shape, scale and width:",
    tuple(row.structure.matrix.shape),
    row.scale,
    row.circuit.n_qubits,
)
print("tridiagonal unnormalized projection:", row.actual.tolist())
print("tridiagonal absolute error:", row.error)
```

## Construct the scaled four by four grid Laplacian

```python
print(models[3].matrix.real.tolist())
```

## Check the five-address Laplacian and compare all projections

```python
row = results[3]
print(
    "laplacian shape, scale and width:",
    tuple(row.structure.matrix.shape),
    row.scale,
    row.circuit.n_qubits,
)
print("laplacian unnormalized projection:", row.actual.tolist())
print("laplacian absolute error:", row.error)
assert all(row.error < 1e-12 for row in results)
display(Image(filename=str(save_projections(results, output))))
```
