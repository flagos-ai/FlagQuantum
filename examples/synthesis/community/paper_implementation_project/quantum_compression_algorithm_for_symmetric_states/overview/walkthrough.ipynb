# Symmetric-state compression

All twelve original source stages, five-qubit uniform input, three-qubit single-excitation input, three-qubit nonsymmetric input and 2048 actual native samples per case.

The single-excitation source lists three equal amplitudes of 1/3. Their squared norm is 1/3, so the native workflow explicitly normalizes those amplitude ratios before preparation. Full absolute complex vectors are compared with an independent Circuit. Native shot seeds are local, with no claim of matching earlier hosted counts.

## Compose the bounded compressor under its symmetry promise

The editable uniform root below prepares five logical qubits and calls the
shared reversible compressor. FlagQuantum composes the supplied local
unitaries, controlled bit operations and logical views. The user supplies
the symmetry promise and the combinatorial rotation construction; this is
not automatic lossy compression of an arbitrary quantum state.

Try the three-bit nonsymmetric input already provided and compare its result
with the symmetric cases. Preserve complete complex amplitudes and the stated
normalization of the source's one-excitation ratios. Each of the three original
cases retains 2048 actual samples; histograms alone do not establish a
compression guarantee outside the promised subspace.


## Load the native runtime

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.community.paper_implementation_project.quantum_compression_algorithm_for_symmetric_states.overview.plots import (
    save_distributions,
)
from examples.synthesis.community.paper_implementation_project.quantum_compression_algorithm_for_symmetric_states.overview.quantum import (
    SOURCE_WEIGHTS,
    adjacent_rotation,
    endpoint_rotation,
    source_amplitudes,
)
from examples.synthesis.community.paper_implementation_project.quantum_compression_algorithm_for_symmetric_states.overview.workflow import (
    SHOTS,
    run_case,
)

output = Path("artifacts/development/symmetric-compression/distributions.png")

from flagquantum import synth
from flagquantum.synth import Output, QArray, QBit, allocate, hadamard_transform, qfunc
from examples.synthesis.community.paper_implementation_project.quantum_compression_algorithm_for_symmetric_states.overview.quantum import (
    compress,
)


@qfunc
def uniform(x: Output[QArray[QBit]]) -> None:
    allocate(5, x)
    hadamard_transform(x)
    compress(5, x)


preview = synth.synthesize(uniform, max_qubits=5, max_workspace=0)
print("Editable symmetric preparation:", preview.report["resources"])
```
