## Keep the starter and the reference answer distinct

The editable starter allocates a logical array and requests uniform
superposition. The separate reference qfunc requests Gaussian probability
loading. FlagQuantum supplies those gates, but the uniform starter is not a
solution to the Gaussian approximation challenge simply because it compiles.
Both original eight-bit and four-bit stages remain available.

Try changing the starter and compare its probability error against the same
signed grid and reference answer. The second stage's width cap of 127 is a
budget, not a request for 127 allocated or simulated qubits. The optional team
export remains disabled, while actual local resource reports and full-state
checks are retained.


Use the native local execution environment.

```python
from pathlib import Path

from IPython.display import Image, display

import examples.synthesis.community.hackathons.iquhack_2025.challenge_template.overview.main as report
import examples.synthesis.community.hackathons.iquhack_2025.challenge_template.overview.workflow as workflow
from flagquantum import synth
from flagquantum.synth import (
    Output,
    QArray,
    QBit,
    allocate,
    hadamard_transform,
    prepare_state,
    qfunc,
)


@qfunc
def starter(width: int, state: Output[QArray[QBit]]) -> None:
    allocate(width, state)
    hadamard_transform(state)


@qfunc
def reference_answer(probabilities: list[float], state: Output[QArray[QBit]]) -> None:
    prepare_state(probabilities, 0, state)


preview = synth.synthesize(
    starter,
    classical_inputs={"width": 8},
    max_qubits=18,
    max_workspace=0,
    basis=("cx", "u"),
    memory_limit_bytes=2**20,
)
print("Editable uniform starter:", preview.report["resources"])
```
