# iQuHack Gaussian reference solution

Run the six-bit direct reference and the original eight- and four-bit amplification stages. Measure the work leakage from the archived overlap and verify the corrected overlap on the same grids.

## Compare a requested state with an amplification construction

The editable direct qfunc supplies normalized Gaussian probabilities to native
state preparation. The amplified route composes the shared load, zero
reflection and checked inverse helpers. FlagQuantum constructs those quantum
operations; the user chooses the grid, overlap estimate and repetition count.
The archived and corrected overlap formulas are deliberately separate.

Try one of the original four- or eight-bit grids and compare conditional state
error, success probability and scratch leakage separately. A normalized
postselected Gaussian does not prove the complete state is clean. Preserve the
archived case as a diagnostic rather than silently correcting its input data.
The direct six-bit loader is a reference route, not evidence that amplification
is cheaper or works unchanged on arbitrary distributions.


Use the local FlagQuantum runtime.

```python
from pathlib import Path

from IPython.display import Image, display

import examples.synthesis.community.hackathons.iquhack_2025.gaussian_reference.overview.main as report
import examples.synthesis.community.hackathons.iquhack_2025.gaussian_reference.overview.workflow as workflow
import torch
import flagquantum as fq
from flagquantum import synth
from flagquantum.synth import Output, QArray, QBit, prepare_state, qfunc


@qfunc
def direct(probabilities: list[float], data: Output[QArray[QBit]]) -> None:
    prepare_state(probabilities, 0, data)


preview = synth.synthesize(
    direct,
    classical_inputs={"probabilities": workflow.target(6).tolist()},
    max_qubits=6,
    max_workspace=0,
)
torch.testing.assert_close(
    fq.run(preview).statevector().flatten(),
    workflow.target(6).sqrt().to(torch.complex128),
    atol=3e-12,
    rtol=0,
)
print("Direct Gaussian resources:", preview.report["resources"])
```
