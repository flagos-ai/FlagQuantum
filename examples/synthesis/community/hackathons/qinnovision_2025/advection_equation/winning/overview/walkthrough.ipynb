custom-rotation and Pauli-product evolution schedules are then built explicitly
with Circuit. Classical RK4, the matrix exponential and phase/sign alignment
are independent host calculations. This workflow is not an automatic compiler
from an advection equation to an optimal quantum algorithm.

unaligned amplitude error from the source's aligned diagnostic, and keep the
original boundary and spatial discretization fixed. The complete 127-term

hardware accuracy or a general claim that one method is always preferable.
