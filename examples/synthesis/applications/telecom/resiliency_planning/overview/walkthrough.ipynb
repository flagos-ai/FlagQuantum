# Network resiliency planning

Run all original-size native layers and inspect actual local outcomes.

## A full-width circuit can still be a one-query experiment

The visible qfunc allocates twenty-four decisions and composes twenty cost/mixer
layers with the source correlated-failure penalty helper. FlagQuantum lowers
those declared phases and named outputs. Host code specifies network demands,
feasibility and the finite-shot objective; it does not ask synth to discover a
resilient network model or its classical optimization strategy.

Try inspecting another forty-angle preview binding before changing penalties.
The retained experiment has one objective query with 20000 objective draws and
20000 final draws. It must not be described as converged optimization. Feasible
sampled assignments, exact bounded references and penalized cost are separate.
Propagate preview algorithm edits to shared main.py for the full workflow.


Load native problem, sampling and plotting helpers.

```python
from flagquantum import synth
from flagquantum.synth import Output
from flagquantum.synth import QArray
from flagquantum.synth import QBit
from flagquantum.synth import RX
from flagquantum.synth import allocate
from flagquantum.synth import hadamard_transform
from flagquantum.synth import qfunc

from examples.synthesis.applications.telecom.resiliency_planning.main import (
    DEFAULT_PARAMS,
    cost_layer,
)


@qfunc
def native_main(params: list[float], z: Output[QArray[QBit]]) -> None:
    allocate(24, z)
    hadamard_transform(z)
    for layer in range(20):
        cost_layer(params[2 * layer], z)
        for bit in range(24):
            RX(params[2 * layer + 1], z[bit])


def build_preview(
    params: tuple[float, ...] = DEFAULT_PARAMS,
) -> synth.Program:
    if len(params) != 40:
        raise ValueError("Twenty resiliency layers require 40 real parameters")
    return synth.synthesize(
        native_main,
        classical_inputs={"params": params},
        max_workspace=0,
        memory_limit_bytes=2**30,
    )


from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.telecom.resiliency_planning.main import (
    build_program,
)
from examples.synthesis.applications.telecom.resiliency_planning.overview import (
    main,
    plots,
    problem,
    workflow,
)
```
