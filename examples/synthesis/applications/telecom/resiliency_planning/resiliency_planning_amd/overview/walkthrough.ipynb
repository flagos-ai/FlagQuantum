## Historical hardware naming does not imply accelerator execution

The visible qfunc retains the same twenty-four decisions and twenty source
cost/mixer layers as the shared resiliency program. FlagQuantum constructs
the declared phase model, while host code chooses angles and interprets network
feasibility. This workflow runs on the native CPU runtime; the AMD folder name
and disabled setup do not establish an accelerator or distributed result.

The original local experiment uses one objective query with 1000 draws and
40000 final draws. Inspect feasibility separately from penalized rank, and do
not label a single evaluated point as optimization convergence. Try another
forty-angle preview binding and propagate algorithm edits to the shared main.py
before rerunning its full host workflow.


Use the installed native runtime. The vendor installation cell becomes an environment note.

```python
from flagquantum import synth
from flagquantum.synth import Output
from flagquantum.synth import QArray
from flagquantum.synth import QBit
from flagquantum.synth import RX
from flagquantum.synth import allocate
from flagquantum.synth import hadamard_transform
from flagquantum.synth import qfunc

from examples.synthesis.applications.telecom.resiliency_planning.main import (
    DEFAULT_PARAMS,
    cost_layer,
)


@qfunc
def native_main(params: list[float], z: Output[QArray[QBit]]) -> None:
    allocate(24, z)
    hadamard_transform(z)
    for layer in range(20):
        cost_layer(params[2 * layer], z)
        for bit in range(24):
            RX(params[2 * layer + 1], z[bit])


def build_preview(
    params: tuple[float, ...] = DEFAULT_PARAMS,
) -> synth.Program:
    if len(params) != 40:
        raise ValueError("Twenty resiliency layers require 40 real parameters")
    return synth.synthesize(
        native_main,
        classical_inputs={"params": params},
        max_workspace=0,
        memory_limit_bytes=2**30,
    )


from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.telecom.resiliency_planning.main import (
    build_program,
)
from examples.synthesis.applications.telecom.resiliency_planning.overview import (
    main as shared_main,
)
from examples.synthesis.applications.telecom.resiliency_planning.overview import (
    plots,
    problem,
    reference,
    workflow,
)
from examples.synthesis.applications.telecom.resiliency_planning.resiliency_planning_amd.overview import (
    main,
)

print("Execution uses the existing FlagQuantum CPU runtime")
```
