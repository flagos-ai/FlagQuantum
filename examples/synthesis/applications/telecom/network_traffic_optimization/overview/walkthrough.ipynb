# Network traffic optimization

Run the full original directed network and inspect actual sampled feasible routes.

## Compile declared flow and congestion phases

The visible qfunc prepares all twenty-four flow decisions and five cost/mixer
layers. Its shared helper preserves three commodity constraints and cubic
congestion terms; FlagQuantum compiles this finite polynomial and named output.
Host problem.py supplies network demands, latency and feasibility, while the
classical optimizer chooses angle bindings from finite-shot objectives.

Inspect valid flow assignments separately from penalized cost. The retained
query budget and 10000-draw readout are observations, not an optimal-routing
certificate. Source MILP objective and actual latency are different quantities.
Propagate preview algorithm edits to shared main.py for the complete search;
a 24-bit simulation does not claim hardware or distributed scalability.


Load the native graph, circuit, optimizer, and plotting helpers.

```python
from flagquantum import synth
from flagquantum.synth import Output
from flagquantum.synth import QArray
from flagquantum.synth import QBit
from flagquantum.synth import RX
from flagquantum.synth import allocate
from flagquantum.synth import hadamard_transform
from flagquantum.synth import qfunc

from examples.synthesis.applications.telecom.network_traffic_optimization.main import (
    DEFAULT_PARAMS,
    cost_layer,
)


@qfunc
def native_main(params: list[float], z: Output[QArray[QBit]]) -> None:
    allocate(24, z)
    hadamard_transform(z)
    for layer in range(5):
        cost_layer(params[2 * layer], z)
        for bit in range(24):
            RX(params[2 * layer + 1], z[bit])


def build_preview(
    params: tuple[float, ...] = DEFAULT_PARAMS,
) -> synth.Program:
    if len(params) != 10:
        raise ValueError("Five traffic layers require ten real parameters")
    return synth.synthesize(
        native_main,
        classical_inputs={"params": params},
        max_workspace=0,
        memory_limit_bytes=2**30,
    )


from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.telecom.network_traffic_optimization.overview import (
    main,
    plots,
    problem,
    workflow,
)
```
