# Full-grid Maxwell evolution

Execute the original 32 by 32 conductor problem through every original-time GQSP walk selection.

Load the native full-grid program and plotting helpers.

```python
from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.physical_systems.maxwell_equation.maxwell_2d_simulation.overview import (
    main,
    phases,
    physics,
    plots,
    quantum,
    workflow,
)
```

Use the source global-phase normalization for field display and a separate complex projection check.

```python
display_fields = plots.fields
```

Retain the original five bits per axis, time five and rectangular conductor.

```python
assert (physics.SIZE, physics.LENGTH, physics.TIME, physics.BOTTOM, physics.TOP) == (5, 32, 5., 9, 24)
```

Keep all four electromagnetic components and both complete spatial axes.

```python
data_register_bits = {"x": 5, "y": 5, "direction": 1, "field": 1}
assert sum(data_register_bits.values()) == 12
```

Compose both exact periodic finite-difference branches.

```python
gradient = quantum.gradient
```

Select forward and backward gradients through the field bit.

```python
gradient_composition = quantum.periodic
```

Retain both Ez/Hx and Ez/Hy couplings and their relative signs.

```python
build_periodic = quantum.periodic
```

Keep the full conductor predicates and restore arbitrary borrowed storage.

```python
conductor = quantum.conductor
```

Fit the original effective-time-640 Laurent polynomial freshly.

```python
fitted = phases.fit()
assert phases.DEGREE == 686 and len(fitted.rows) == 1373
```

Use the original three physical field panels and per-field display normalization.

```python
figure_directory = Path("maxwell-figures")
figure_directory.mkdir(exist_ok=True)
```

Generate every original initial amplitude and display the initial fields.

```python
initial = physics.initial_state()
assert initial.numel() == 4096
plots.fields(initial, "Original full-grid initial state at t=0", figure_directory / "initial-fields.png")
display(Image(filename=str(figure_directory / "initial-fields.png")))
```

Retain all seventeen physical bits including the GQSP auxiliary.

```python
run_simulation = workflow.run
```

Use the bounded native CPU execution loop for the complete source sequence.

```python
simulate = run_simulation
```

Execute all 1372 original walk selections on the full grid.

```python
result = simulate(fitted)
```

Read the actual block-zero amplitudes directly.

```python
print(tuple(complex(value) for value in result.projected[:8]))
```

Expose the source y*32+x indexing through a labeled field tensor.

```python
fields = result.projected.reshape(4, 32, 32)
assert fields.shape == (4, 32, 32)
```

Display actual native fields at the original time five.

```python
filenames = plots.save(result, figure_directory)
display(Image(filename=str(figure_directory / "evolved-fields.png")))
```

Use the independently checked full-grid classical generator and exponential action.

```python
classical = result.classical
assert classical.numel() == 4096
```

Check the original fidelity and amplitude scale, plus the absolute complex error.

```python
main.report(result)
assert 1 - result.fidelity < 1e-4
assert result.absolute_error < 1e-6
main.write(result, Path("maxwell-results.json"))
```

Display the original classical and quantum field comparisons.

```python
for filename in ("classical-fields.png", "quantum-fields.png"):
    display(Image(filename=str(figure_directory / filename)))
```
