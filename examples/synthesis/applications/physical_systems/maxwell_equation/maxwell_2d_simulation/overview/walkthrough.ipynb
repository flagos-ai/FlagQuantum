# Full-grid Maxwell evolution

Execute the original 32 by 32 conductor problem through every original-time GQSP walk selection.

## The field encoding is explicit Circuit composition

The displayed block factory composes conductor predicates and periodic
finite-difference helpers on the original twelve data bits plus block storage.
This workflow uses FlagQuantum Circuit kernels and local execution; it does
not yet express the complete field encoding and phase sequence as a native
qfunc. Host physics.py defines field layout, the conductor interval and grid.

Inspect the 32-by-32 field components and unnormalized projection before
comparing fidelity. All 1373 phase rows for degree 686 remain; phase fitting,
classical PDE reference and quantum projection are separate computations.
Changing conductor geometry requires matching both models and borrowed-bit
checks. A low normalized state error alone does not determine physical field
normalization or postselection cost.


Load the native full-grid program and plotting helpers.

```python
import flagquantum as fq
import torch

from examples.synthesis.applications.physical_systems.maxwell_equation.maxwell_2d_simulation.overview.quantum import (
    conductor,
    periodic,
)


def block_preview() -> fq.Circuit:
    circuit = fq.Circuit(16, dtype=torch.complex128)
    conductor(circuit)
    periodic(circuit)
    conductor(circuit)
    return circuit


from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.physical_systems.maxwell_equation.maxwell_2d_simulation.overview import (
    main,
    phases,
    physics,
    plots,
    quantum,
    workflow,
)
```

```python
result = simulate(fitted)
```

Check the original fidelity and amplitude scale, plus the absolute complex error.

```python
main.report(result)
assert 1 - result.fidelity < 1e-4
assert result.absolute_error < 1e-6
main.write(result, Path("maxwell-results.json"))
```
