## Distinguish the native recurrence from Circuit demonstrations

The visible native root composes bounded, explicitly quantized Brownian/
Chebyshev helpers and returns named time and price fields. FlagQuantum builds
the arithmetic and lifecycle operations for this retained finite recurrence.
The standalone signed sine and recurrence demonstrations in quantum.py use
explicit Circuit XOR tables; those demonstrations are not evidence that the
same signed operations are automatically optimized by synth.

Inspect nonzero time/price probabilities and fixed-point scale before changing
coefficients. Quantization error, private dropped registers and compiler
workspace have different meanings. The independent full-state check preserves
all 32 nonzero words; a finite coherent recurrence does not establish exact
continuous Brownian dynamics or a scalable pricing algorithm.


Regenerate the truncated Gaussian distribution by integrating the bin masses.

```python
from flagquantum import synth
from flagquantum.synth import Output
from flagquantum.synth import QNum
from flagquantum.synth import allocate
from flagquantum.synth import drop
from flagquantum.synth import hadamard_transform
from flagquantum.synth import qfunc

from examples.synthesis.applications.finance.brownian_chebyshev_polynomials.main import (
    return_to_price_space,
    series,
)


@qfunc
def native_main(t: Output[QNum], price: Output[QNum]) -> None:
    b: QNum = QNum()
    allocate(1, t)
    hadamard_transform(t)
    series(t, b)
    return_to_price_space(b, t, price)
    drop(b)


def build_preview() -> synth.Program:
    return synth.synthesize(
        native_main,
        max_qubits=23,
        max_workspace=2,
        max_operations=3000,
        memory_limit_bytes=256 * 1024**2,
    )


from pathlib import Path

from IPython.display import Image, display

from examples.synthesis.applications.finance.brownian_chebyshev_polynomials.overview import (
    model,
    workflow,
)
from examples.synthesis.applications.finance.brownian_chebyshev_polynomials.overview.main import (
    print_report,
    save_figures,
)
```
