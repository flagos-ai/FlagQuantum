# Fresh original-sized partial exponential preparation

Native high-level interval preparation with exact fresh overlap, independent full-state checks and explicit source-error diagnostics. No original source parameter is silently repaired or accepted.


## Stage 1

Fresh original five-bit exponential state using existing native amplitude preparation.


```python
import json
from dataclasses import asdict
from pathlib import Path

from IPython.display import display
from PIL import Image

from examples.synthesis.applications.finance.autocallable_options.partial_exponential_state_preparation.overview.model import (
    make_case,
    source_cases,
)
from examples.synthesis.applications.finance.autocallable_options.partial_exponential_state_preparation.overview.plots import (
    plot_amplitudes,
)
from examples.synthesis.applications.finance.autocallable_options.partial_exponential_state_preparation.overview.workflow import (
    prepare_distribution,
    prepare_interval,
)

EXP_RATE, NUM_QUBITS = 0.5, 5
first_case, shifted_case = source_cases(EXP_RATE)
base = prepare_distribution(first_case.amplitudes)
print("Original problem bits:", NUM_QUBITS)
print("Actual base resources:", base.resources)
print("Independent base check:", base.check)
```

## Stage 2

Read x by its named register and plot all32words without bitstring decoding.


```python
output = Path("artifacts/development/partial_exponential_state_preparation")
base_figure = plot_amplitudes(base.probabilities, output, "full-exponential")
with Image.open(base_figure) as figure:
    display(figure.copy())
```

## Stage 3

Retain the original approximate overlap formula and independently prove the discrete overlap.


```python
X0, X1 = 15, 29
print("Source continuous-overlap hint:", first_case.declared_overlap)
print("Fresh proven discrete overlap:", first_case.observed_overlap)
try:
    make_case(EXP_RATE, X0, X1, supplied_overlap=first_case.declared_overlap)
except ValueError as error:
    print("Original exact-amplification hint remains refused:", error)
else:
    raise AssertionError("the source mismatching overlap must not be accepted")
print("Native interval preparation derives the exact overlap from fresh amplitudes.")
print(
    "This follows the existing approved correction policy and adds no unchanged-original coverage."
)
```

## Stage 4

Run source-sized first-interval amplification with exact overlap and independent full physical-state validation.


```python
first = prepare_interval(first_case)
print("First interval resources:", first.resources)
print("First interval state, workspace and attenuation checks:", first.check)
first_figure = plot_amplitudes(
    first.probabilities, output, "first-interval", interval=(X0, X1)
)
with Image.open(first_figure) as figure:
    display(figure.copy())
```

## Stage 5

Compute the small target interval hint without running an unrequested long quantum circuit.


```python
X0, X1 = 3, 13
small_case = make_case(EXP_RATE, X0, X1)
print("Source small-interval hint:", small_case.declared_overlap)
print("Discrete small-interval overlap:", small_case.observed_overlap)
print("Overlap-derived iteration count:", small_case.iterations)
```

## Stage 6

Prepare the original upper interval, shift by minus18modulo32 and retain all output words.


```python
print("Original shifted source hint:", shifted_case.declared_overlap)
print("Fresh proven shifted overlap:", shifted_case.observed_overlap)
try:
    make_case(EXP_RATE, 21, 31, -18, supplied_overlap=shifted_case.declared_overlap)
except ValueError as error:
    print("Original final exact-amplification hint remains refused:", error)
else:
    raise AssertionError("the final source hint must not bypass its semantic refusal")
shifted = prepare_interval(shifted_case)
print("Shifted resources:", shifted.resources)
print("Shifted absolute-state, workspace and attenuation checks:", shifted.check)
print(
    "Small versus shifted overlap-derived iteration counts:",
    small_case.iterations,
    shifted_case.iterations,
)
shifted_figure = plot_amplitudes(
    shifted.probabilities, output, "shifted-interval", interval=(X0, X1)
)
with Image.open(shifted_figure) as figure:
    display(figure.copy())
```

## Stage 7

Retain the source0.01amplitude tolerance and separately prove clean retained attenuation and compiler work.


```python
for prepared in (first, shifted):
    assert prepared.check.source_amplitude_error < 0.01
    assert prepared.check.state_error < 1e-11
    assert prepared.check.workspace_error < 1e-11
    assert prepared.check.private_attenuation_one_probability is not None
    assert prepared.check.private_attenuation_one_probability < 1e-22
    assert prepared.check.interval_leakage < 1e-22
print(
    "Attenuation is retained private storage;its proved-zero state does not reclassify it as compiler workspace."
)
print(
    "ARTIFACT_JSON="
    + json.dumps(
        {
            "problem_bits": NUM_QUBITS,
            "rate": EXP_RATE,
            "first_case": asdict(first_case),
            "small_case": asdict(small_case),
            "shifted_case": asdict(shifted_case),
            "base": {"check": asdict(base.check), "resources": asdict(base.resources)},
            "first": {
                "check": asdict(first.check),
                "resources": asdict(first.resources),
            },
            "shifted": {
                "check": asdict(shifted.check),
                "resources": asdict(shifted.resources),
            },
            "semantic_adaptation": "Original invalid overlap hints remain refused;native exact overlap is freshly derived under the existing approved correction policy.",
        }
    )
)
```
