## Original code cell 2


```python
from io import BytesIO

from IPython.display import Image, display
from matplotlib import pyplot as plt
from matplotlib.figure import Figure

from examples.synthesis.applications.cybersecurity.patching_management.patch_min_vertex_cover.comparisons.circuit import (
    BINDINGS,
)
from examples.synthesis.applications.cybersecurity.patching_management.patch_min_vertex_cover.overview.operations import (
    verify,
)
from examples.synthesis.applications.cybersecurity.patching_management.patch_min_vertex_cover.overview.workflow import (
    build_program,
    dual_graph,
    kill_chain_graph,
    model_tables,
    plot_costs,
    plot_graph,
    plot_trace,
    remaining_graph,
    run_workflow,
    source_graph,
)


def show_figure(figure: Figure) -> None:
    buffer = BytesIO()
    figure.savefig(buffer, format="png", dpi=130)
    display(Image(buffer.getvalue()))
    plt.close(figure)


graph = source_graph()
print("Original resources:", tuple(range(8)))
print("Original vulnerabilities:", tuple("ABCDEFG"))
print("Original incidences:", graph.number_of_edges())
show_figure(
    plot_graph(graph, "Original resource-vulnerability graph", layout="bipartite")
)
```
