# Whitebox fuzzing with quantum search

Preserve the classical branching program and six-bit x/y search domain. The quantum oracle marks exactly branch c. Classical sampling preserves the source seed 3 and 500 draws; quantum sampling draws 500 local observations from the checked native state. This is a local algorithm demonstration, not a speedup or hardware-fuzzing claim.


```python
from io import BytesIO

from IPython.display import Image, display
from matplotlib import pyplot as plt

from examples.synthesis.applications.cybersecurity.whitebox_fuzzing.overview.operations import (
    build_program,
)
from examples.synthesis.applications.cybersecurity.whitebox_fuzzing.overview.workflow import (
    classical_samples,
    foo,
    marked_pairs,
    output_counts,
    plot_outputs,
    quantum_samples,
    source_recommendation,
    verify,
)

print(foo(12, 4), foo(12, 3), foo(1, 1), foo(0, 0), foo(63, 63))
```

```python
classical = classical_samples()
print("Classical samples:", sum(row.count for row in classical))
print(output_counts(classical))
classical_figure = plot_outputs(classical, "Classical whitebox sampling: 500 draws")
classical_png = BytesIO()
classical_figure.savefig(classical_png, format="png", dpi=130)
display(Image(classical_png.getvalue()))
plt.close(classical_figure)
```

## Quantum oracle

The existing native helper expresses `(x+y<9) & ((x%4)*(y%2)==1)` and a conditional pi phase. Two named six-bit registers preserve the source structure fields. Enumerating the classical predicate gives six marked pairs, all reaching branch c.


```python
print("Marked pairs:", marked_pairs())
assert all(foo(x, y) == "c" for x, y in marked_pairs())
```

```python
oracle = build_program(None)
print(oracle.report["resources"])
oracle_check = verify(oracle, None)
print(oracle_check)
```

## Original recommendation and actual program

The source estimate assumes two marked inputs and rounds to 36 iterations, but the actual source search runs 10 iterations. Preserve both. The actual predicate marks six pairs; these facts describe different quantities and do not establish that either iteration count is optimal.


```python
print("Source heuristic recommendation:", source_recommendation())
print("Actual source search rounds:", 10)
print("Actual marked pairs:", len(marked_pairs()))
```

```python
program = build_program(10)
```

```python
print("Width ceiling:", 25, "Actual resources:", program.report["resources"])
```

```python
checks = tuple(verify(build_program(r), r) for r in (0, 1, 2, 10))
print("Independent full-state checks:", checks)
```

```python
observations = quantum_samples(program)
assert sum(row.count for row in observations) == 500
print("Actual quantum sample count:", sum(row.count for row in observations))
```

```python
print("x, y, counts, exact probability, foo output")
for row in sorted(observations, key=lambda row: (-row.count, row.x, row.y))[:20]:
    print(row)
print("Quantum branch counts:", output_counts(observations))
```

```python
quantum_figure = plot_outputs(
    observations, "Quantum whitebox sampling: 500 draws, 10 Grover rounds"
)
quantum_png = BytesIO()
quantum_figure.savefig(quantum_png, format="png", dpi=130)
display(Image(quantum_png.getvalue()))
plt.close(quantum_figure)
```
