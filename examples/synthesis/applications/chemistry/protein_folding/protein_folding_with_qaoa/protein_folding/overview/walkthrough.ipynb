## Model inputs


```python
from collections import Counter

from IPython.display import Image, display

from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_qaoa.protein_folding.overview.main import (
    OUTPUT,
    report,
)
from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_qaoa.protein_folding.overview.model import (
    RESIDUES,
    WIDTH,
    costs,
    directions,
    lattice,
)
from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_qaoa.protein_folding.overview.plots import (
    save_figures,
)
from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_qaoa.protein_folding.overview.quantum import (
    LAYERS,
    build_program,
)
from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_qaoa.protein_folding.overview.workflow import (
    run_workflow,
)
```

## Classical objective


```python
energies = costs()
print("Assignments and cost range:", len(energies), min(energies), max(energies))
```

## The original six-residue chain


```python
print("Residues, bits and layers:", RESIDUES, WIDTH, LAYERS)
```

## Native QAOA


```python
probe = build_program((0.01,) * (2 * LAYERS))
print("Logical ports:", probe.registers)
```

## Resource inspection


```python
print("Resources:", probe.report["resources"])
```

## Run the finite-shot CVaR optimization


```python
result = run_workflow()
report(result)
```

## Optimization trace


```python
figures = save_figures(result, OUTPUT)
display(Image(filename=str(figures[0])))
```

## Five lowest-cost observed assignments


```python
frequencies = Counter(result.optimized.tolist())
ordered = sorted(frequencies, key=lambda word: (energies[word], word))
for word in ordered[:5]:
    print(word, energies[word], frequencies[word], directions(word))
```
