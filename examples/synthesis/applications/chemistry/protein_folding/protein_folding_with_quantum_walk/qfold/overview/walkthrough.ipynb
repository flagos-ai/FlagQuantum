# QFold quantum walk

Derive the original coin angles from the archived alanylalanine input data, then run and verify the one-step native quantum walk.

## Compose coin, shift and inverse coin on logical data

The visible qfunc prepares two logical angle bits, composes the checked coin
inside within/apply, shifts the selected move and applies the oracle.
FlagQuantum constructs inverse composition and named logical readout. Host
code reads the archived finite energy differences and derives nonzero coin
angles; it does not ask synth to discover a protein-walk algorithm.

Try another admitted beta in the existing angle calculation and inspect the
full logical distribution. Private dropped coin and move registers remain
retained environment, not automatically zeroed compiler work bits. The
source's 01-prefix energy lookup and sampled counts retain their own bit
convention. This one-pair example is not evidence of scalable protein folding.


Load the native workflow.

```python
from flagquantum import synth
from flagquantum.synth import Output
from flagquantum.synth import QBit
from flagquantum.synth import allocate
from flagquantum.synth import apply
from flagquantum.synth import drop
from flagquantum.synth import inplace_prepare_amplitudes
from flagquantum.synth import qfunc
from flagquantum.synth import within

from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_quantum_walk.qfold.main import (
    AMPLITUDES,
    ANGLES,
    coin,
    oracle,
    shift,
)


@qfunc
def native_main(
    a: float, b: float, amplitudes: list[float], phi: Output[QBit], psi: Output[QBit]
) -> None:
    coin_bit: QBit = QBit()
    move_bit: QBit = QBit()
    allocate(coin_bit)
    allocate(move_bit)
    allocate(phi)
    allocate(psi)
    inplace_prepare_amplitudes(amplitudes, 0, (psi, phi))
    with within():
        coin(a, b, coin_bit, move_bit, psi, phi)
    with apply():
        shift(coin_bit, move_bit, psi, phi)
    oracle(coin_bit, move_bit)
    drop(coin_bit)
    drop(move_bit)


def build_preview(
    a: float = ANGLES[0],
    b: float = ANGLES[1],
    amplitudes: tuple[float, ...] = AMPLITUDES,
) -> synth.Program:
    if len(amplitudes) != 4:
        raise ValueError("Use four real logical-data amplitudes")
    return synth.synthesize(
        native_main, classical_inputs={"a": a, "b": b, "amplitudes": list(amplitudes)}
    )


from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_quantum_walk.qfold.overview import (
    main,
    workflow,
)
```

Load the original alanylalanine input data.

```python
data = workflow.dataset()
```

```python
print(f"Energy differences: {data.energy_differences}")
```

Read the original dihedral angles.

```python
phi_angles, psi_angles = data.phi_angles, data.psi_angles
```

Check both original one-angle registers.

```python
assert len(phi_angles) == len(psi_angles) == 1
```

Compute the move and rotation register widths.

```python
main.describe(data)
```

Use the Metropolis acceptance formula to derive both angles.

```python
beta = 1.0
```

Evaluate both coin rotations using the original move assignment.

```python
angles = workflow.coin_angles(data, beta)
print(f"Coin rotations by move (radians): {angles}")
```

Build the native one-step walk from the two calculated rotations.

```python
program = build_preview(*angles)
```

Inspect the synthesized circuits and independent complete-state check.

```python
main.verify(program, angles)
```

Draw the original 2,048 samples from the two logical outputs.

```python
counts = workflow.sample(program)
```

Read back the observed counts.

```python
main.print_counts(counts)
```
