# QFold quantum walk

Derive the original coin angles from the archived alanylalanine input data, then run and verify the one-step native quantum walk.

Load the native workflow.

```python
from examples.synthesis.applications.chemistry.protein_folding.protein_folding_with_quantum_walk.qfold.overview import main, workflow
```

Load the original alanylalanine input data.

```python
data = workflow.dataset()
```

Inspect all sixteen energy differences.

```python
print(f"Energy differences: {data.energy_differences}")
```

Read the original dihedral angles.

```python
phi_angles, psi_angles = data.phi_angles, data.psi_angles
```

Check both original one-angle registers.

```python
assert len(phi_angles) == len(psi_angles) == 1
```

Compute the move and rotation register widths.

```python
main.describe(data)
```

Use the Metropolis acceptance formula to derive both angles.

```python
beta = 1.0
```

Evaluate both coin rotations using the original move assignment.

```python
angles = workflow.coin_angles(data, beta)
print(f"Coin rotations by move (radians): {angles}")
```

Build the native one-step walk from the two calculated rotations.

```python
program = workflow.build(angles)
```

Inspect the synthesized circuits and independent complete-state check.

```python
main.verify(program, angles)
```

Draw the original 2,048 samples from the two logical outputs.

```python
counts = workflow.sample(program)
```

Read back the observed counts.

```python
main.print_counts(counts)
```

```python
print(f"Source 01-prefix energy lookup: {workflow.energy_lookup(data)}")
```
