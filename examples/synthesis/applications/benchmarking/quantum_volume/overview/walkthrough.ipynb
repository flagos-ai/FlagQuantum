# Complete native random-circuit volume experiment

Retain every original local experiment: sizes 3 through 6, as many layers as
qubits, ten fresh random circuits per size and 100 actual categorical draws per
circuit. All forty synthesized programs are retained. Each two-qubit matrix is
fresh; no stored Qmod matrix, importer, remote job or fixed-gate substitution is
used. Native qfunc helpers capture immutable matrices and compose before
synthesis; no Python random function executes inside captured quantum code.

The source uses the median of **positive observed counts**, rather than the
median of ideal output probabilities. We preserve its score, standard-deviation
formula and first-failure decision, explicitly label this a source diagnostic,
and show the standard ideal heavy mass separately. The result is not a standard
quantum-volume certification or a noisy-device capability claim.

The two independent Haar factors use NumPy complex Gaussian QR instead of SciPy,
with the same left-invariant distribution and subsequent source QR construction.
Python permutation and NumPy matrix generators are seeded independently at zero.
Exact random-number equivalence to the source SciPy version is not asserted.
Readout uses independent explicit seeds and samples the native named marginal.

```python
print("Complete experiment:", run_experiment.__name__)
print("Source deviation = population std of ten heavy counts / (10 * 100)")
print("Decision: score - 2 * deviation >= 2/3; stop on the first failed size")
```

```python
num_trials = 10
num_shots = 100
min_qubits = 3
max_qubits = 6
experiment = run_experiment()
qprogs = experiment.programs
assert len(qprogs) == 40
assert (
    sum(sum(count for _, count in trial.counts) for trial in experiment.trials) == 4000
)
for row in experiment.sizes:
    print(
        row.size,
        row.observed_heavy_probability,
        source_round_significant(row.source_standard_deviation),
        row.lower_bound,
        row.passes_source_threshold,
    )
print("Source observed-count volume diagnostic:", experiment.source_volume)
print(
    "Maximum absolute full-state error:", max(t.state_error for t in experiment.trials)
)
print(
    "Maximum named probability error:",
    max(t.named_probability_error for t in experiment.trials),
)
print(
    "Actual native CX counts:",
    tuple(
        (t.size, t.index, t.resources.operation_counts.get("cx", 0))
        for t in experiment.trials
    ),
)
output_directory = Path("artifacts/development/quantum_volume")
plot_experiment(experiment, output_directory)
figure_path = output_directory / "observed_and_ideal_heavy_outputs.png"
display(Image(filename=str(figure_path)))
```
