# Register addition

Native adaptation of the frozen addition catalog. Signature tables retain widths, directions and encodings. Resource identities use native names. These are declared models. Run `python -m examples.algorithms.classical_register_addition` for the executable `circuit.append(op)` workflow.

Copyright 2023 Google LLC. Apache-2.0. Source and blob are recorded in `ADDITION_FILE_AUDIT.md`.


```python
from IPython.display import SVG, display
from examples.algorithms.addition_catalog import (
    addition_examples, addition_example_graph, addition_signature_svg,
)
from flagquantum.drawer.call_graph import export_call_graph_dot, render_call_graph_svg

catalog = {name: (name, call, signature) for name, call, signature in addition_examples()}

from flagquantum.drawer.addition_symbols import addition_circuit_labels
```

## In-place addition

Computes a+b modulo the result width, preserving a. The source ripple uses n−1 AND computations and measured erasures. Symbolic models and concrete primitive emission are separate accounts.


```python
from examples.algorithms.addition_catalog import addition_signature_data
from flagquantum.drawer.register_score_mpl import draw_register_score
import matplotlib.pyplot as plt

figure, axes = draw_register_score(addition_signature_data(catalog["add_small"]), unit_to_inches=1.6)
plt.show()
```

### Example Instances

```python
add_symb = catalog["add_symb"]
```

```python
add_small = catalog["add_small"]
print(addition_circuit_labels(4, 4))
```

```python
add_large = catalog["add_large"]
```

```python
add_diff_size_regs = catalog["add_diff_size_regs"]
```

#### Graphical Signature

```python
display(SVG(addition_signature_svg([add_symb, add_small, add_large, add_diff_size_regs])))
```

### Call Graph

```python
add_symb_g = addition_example_graph(add_symb[1])
add_symb_sigma = add_symb_g.leaf_counts
display(SVG(render_call_graph_svg(export_call_graph_dot(add_symb_g.root, add_symb_g.direct_calls))))
print(add_symb_sigma)
```

## Out-of-place addition

Computes a+b into a separate output with one carry bit. The full source loop uses n ANDs, costing 4n T gates. Its truncated variant uses n−1. The source prose states 4n−4 without distinguishing this flag. Measured adjoint consumes a correct sum with zero T gates.


```python
from examples.algorithms.addition_catalog import addition_signature_data
from flagquantum.drawer.register_score_mpl import draw_register_score
import matplotlib.pyplot as plt

figure, axes = draw_register_score(addition_signature_data(catalog["add_oop_small"]), unit_to_inches=1.6)
plt.show()
```

### Example Instances

```python
add_oop_symb = catalog["add_oop_symb"]
```

```python
add_oop_small = catalog["add_oop_small"]
```

```python
add_oop_large = catalog["add_oop_large"]
```

#### Graphical Signature

### Call Graph
