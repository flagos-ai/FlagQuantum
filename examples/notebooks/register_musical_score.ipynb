# Register musical score

The component dependency graph can be laid out as a musical score. Time runs
from left to right, and horizontal lines represent qubits or registers.

Adapted from Qualtran v0.7.0, `qualtran/drawing/musical_score.ipynb`, commit
`bcb17bb25360ed68ce3bc084645db3e929799389`.
Copyright 2023 The Qualtran Authors. Licensed under Apache-2.0.


## Graph

Render the shared four-control AND ladder with control values `(1, 1, 0, 1)`.
The third control has negative polarity. Logical prefix outputs belong to the
caller. Drawing adds no physical operations or work qubits.


```python
from graphviz import Source
from IPython.display import SVG, display

from flagquantum.drawer.register_graph import export_register_graph_dot
from flagquantum.drawer.register_score import multi_and_register_score

score = multi_and_register_score((1, 1, 0, 1))
dot = export_register_graph_dot(score)
display(SVG(Source(dot).pipe(format="svg")))
```

## Matplotlib musical score

Build score data from the same native layout and render it with Matplotlib.
Positions, wire lifetimes and symbols follow the locked source. Native port
identifiers describe logical register lanes instead of foreign object identities.
The caller closes the displayed figure to release its resources.


```python
import matplotlib.pyplot as plt

from flagquantum.drawer.register_score_data import register_score_data
from flagquantum.drawer.register_score_mpl import draw_register_score

msd = register_score_data(score)
fig, ax = draw_register_score(msd)
fig.tight_layout()
display(fig)
plt.close(fig)
```

## Javascript musical score

Export the drawing data to `musical_score_example.json` in the current directory.
The native self-contained HTML viewer embeds that same data and supports both
sequential and dependency views. It replaces the source's separate D3 page and
script. The JSON is drawing metadata, with no executable gate-count model.


```python
from pathlib import Path

from flagquantum.drawer.register_score_data import export_register_score_json
from flagquantum.drawer.register_score_html import export_register_score_html

Path("musical_score_example.json").write_text(
    export_register_score_json(msd), encoding="utf-8"
)
Path("musical_score_example.html").write_text(
    export_register_score_html(msd), encoding="utf-8"
)
```
