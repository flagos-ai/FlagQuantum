# Native register graph drawing

Ported from Qualtran v0.7.0 graphviz.ipynb (Apache-2.0, Copyright 2023 Google LLC). Source commit bcb17bb25360ed68ce3bc084645db3e929799389. Native register graphs preserve dependencies and word widths; metadata-only signatures do not promise executable operations.

## Basic usage

```python
from IPython.display import SVG, display
from flagquantum.core.ir.program.registers import Register, RegisterSide, RegisterType
from flagquantum.core.ir.program.signatures import Signature
from flagquantum.drawer.signature_graph import export_signature_graph_dot
from flagquantum.drawer.call_graph import render_call_graph_svg
from examples.algorithms.register_graphs import composite_dot, classical_predicate_dot

renders = []
def show_dot(dot):
    svg = render_call_graph_svg(dot)
    renders.append(svg)
    display(SVG(svg))
```

Each instance is a register table. Edges show dependencies and word widths. An atomic display is distinct from expanding its real program body.

```python
atom_signature = Signature((Register("q", RegisterType("bit", 1)),))
show_dot(export_signature_graph_dot(atom_signature, "Atom", pretty=True))
```

```python
parallel_signature = Signature((Register("reg", RegisterType("opaque", 3)),))
show_dot(export_signature_graph_dot(parallel_signature, "Parallel X", pretty=True))
```

```python
dot = composite_dot(pretty=True)
show_dot(dot)
```

### Register groups

Preserve the source 100/200-bit words and LEFT/RIGHT arrays. Opposite-side names use distinct native lifetime identities (out: prefix on collisions). Shapes and widths stay intact; these signatures are drawing metadata and are not simulated.

```python
def signature_dot(registers):
    return export_signature_graph_dot(Signature(registers), "Signature", pretty=True)
```

```python
registers = [Register("x", RegisterType("opaque", 100)), Register("y", RegisterType("opaque", 200))]
show_dot(signature_dot(registers))
```

```python
registers = [Register("x", RegisterType("opaque", 100)), Register("a", RegisterType("opaque", 2), (2,), RegisterSide.LEFT), Register("y", RegisterType("opaque", 200))]
show_dot(signature_dot(registers))
```

```python
registers = [Register("x", RegisterType("opaque", 100)), Register("a", RegisterType("opaque", 2), (2, 2), RegisterSide.LEFT), Register("a", RegisterType("opaque", 8), side=RegisterSide.RIGHT), Register("y", RegisterType("opaque", 200))]
show_dot(signature_dot(registers))
```

```python
registers = [Register("x", RegisterType("opaque", 100)), Register("a", RegisterType("opaque", 2), (2, 2), RegisterSide.LEFT), Register("a", RegisterType("opaque", 4), (2,), RegisterSide.RIGHT), Register("y", RegisterType("opaque", 200))]
show_dot(signature_dot(registers))
```
