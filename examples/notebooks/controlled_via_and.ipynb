Adapted from Qualtran v0.7.0, Copyright 2024 Google LLC, Apache-2.0. The signed controls use the native big-endian two’s-complement wire binding. Calls in the graphs are source resource declarations. The append example reports emitted instructions separately.

```python
from collections import Counter

from IPython.display import SVG, display

from examples.algorithms.field_addition_catalog import field_signature_svg
from flagquantum.drawer.call_graph import export_call_graph_dot, render_call_graph_svg
from flagquantum.dynamic import DynamicCircuit


def show_calls(graph):
    display(SVG(render_call_graph_svg(export_call_graph_dot(graph.root, graph.direct_calls))))
    display(graph.leaf_counts)
```

## `ControlledViaAnd`
Reduces a generic controlled bloq to a singly-controlled bloq using an And ladder.

Implements a generic controlled version of the subbloq, by first reducing the
arbitrary control to a single qubit, and then using a single-qubit-controlled
variant of the subbloq.

For signature, see :class:`Controlled`.

#### Parameters
 - `subbloq`: The bloq we are controlling.
 - `ctrl_spec`: The specification for how to control the bloq.


```python
from examples.algorithms.controlled_via_and import ControlledHadamardExample
```

### Example Instances

```python
controlled_via_and_ints = ControlledHadamardExample("integers")
```

```python
controlled_via_and_qbits = ControlledHadamardExample("bits")
```

#### Graphical Signature

```python
display(SVG(field_signature_svg([
    ("controlled_via_and_ints", controlled_via_and_ints.signature, {}),
    ("controlled_via_and_qbits", controlled_via_and_qbits.signature, {}),
])))
```

### Call Graph

```python
show_calls(controlled_via_and_ints.call_graph(depth=1))
```

## Nested Controls
Calling `controlled` on a `ControlledViaAnd` returns another `ControlledViaAnd` by combining the existing and new controls into a single control specification.

```python
nested = ControlledHadamardExample("nested")
display(SVG(field_signature_svg([("nested control", nested.signature, {})])))

circuit = DynamicCircuit(12).append(nested)
display(Counter(gate.name for gate in circuit.to_ir().instructions))
```

```python
show_calls(nested.call_graph())
```
