This notebook reuses native class discovery and reference parsing for all public classes owned by the algorithms package. Each domain retains its own class hierarchy. Reference text comes from existing docstrings.


```python
from tools.development.component_discovery import get_component_classes
component_classes = get_component_classes(is_component=lambda cls: True)
len(component_classes)
```

```python
from tools.development.docstrings import get_references, Reference, UnparsedReference
references = {component_cls: get_references(component_cls) for component_cls in component_classes}
```

```python
from IPython.display import Markdown

def show_references(component_cls, refs):
    if not refs:
        return
    
    text = f'### {component_cls.__name__}\n'
    for ref in refs:
        text += f' - {ref.text}\n'
    
    display(Markdown(text))
```

```python
for component_cls in component_classes:
    show_references(component_cls, references[component_cls])
```

## Parsing Errors

```python
for component_cls, refs in references.items():
    for ref in refs:
        if isinstance(ref, UnparsedReference):
            print(component_cls.__name__)
            print(ref.text)
```

## Backreference

```python
from collections import defaultdict
backrefs = defaultdict(list)

for component_cls, refs in references.items():
    for ref in refs:
        if not hasattr(ref, 'url'):
            continue

        backrefs[ref.url].append(component_cls)
backrefs = dict(backrefs)
```

```python
# TODO: check for disagreement
titles = {ref.url: ref.title for refs in references.values() for ref in refs if hasattr(ref, 'url')}
```

```python
for url, clss in sorted(backrefs.items(), key=lambda x: -len(x[1])):
    text = f'### [{titles[url]}]({url})\n'
    class_names = [f'{cls.__module__}.{cls.__qualname__}' for cls in clss]
    for cn in sorted(class_names):
        text += f' - {cn}\n'

    display(Markdown(text))
```
