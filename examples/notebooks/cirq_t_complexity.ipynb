# Legacy T complexity

Adapted from Qualtran v0.7.0. The shared compiler model reports declared T, Clifford and rotation categories.

```python
from io import BytesIO
from IPython.display import Image, display
import matplotlib.pyplot as plt
from flagquantum.compiler._component_trace.resource_generalization import ConjunctionCall
from flagquantum.ecosystem.cirq._legacy_t_complexity import legacy_compiler_t_complexity
```

## Two-control conjunction

The source conjunction declares four T gates. The legacy conversion also folds the source Clifford overhead into the result.

```python
cost = legacy_compiler_t_complexity(ConjunctionCall((1, 1)))
assert (cost.t, cost.clifford, cost.rotations) == (4, 9, 0)
display(cost.asdict())
```

## Adjoint of the two-control conjunction

The measurement-assisted adjoint declares zero T gates. Its Clifford-equivalent total includes the measurement under the legacy convention.

```python
adjoint_cost = legacy_compiler_t_complexity(ConjunctionCall((1, 1), adjoint=True))
assert (adjoint_cost.t, adjoint_cost.clifford, adjoint_cost.rotations) == (0, 4, 0)
display(adjoint_cost.asdict())
```

## Conjunction with multiple controls

The ladder has one conjunction for each control after the first.

```python
n = 5
display(legacy_compiler_t_complexity(ConjunctionCall((1,) * n)).asdict())
```

```python
def conjunction_t_sweep(n_max=10):
    # Preserve the original example's final point at n_max + 1.
    widths = list(range(3, n_max + 2))
    return widths, [legacy_compiler_t_complexity(ConjunctionCall((1,) * n)).t for n in widths]

widths, counts = conjunction_t_sweep()
assert counts == [4 * (n - 1) for n in widths]
```

```python
fig, ax = plt.subplots()
ax.plot(widths, counts, "o-", label="T")
ax.set(xlabel="Control qubits", ylabel="Gate count", title="Conjunction gate")
ax.legend()
buffer = BytesIO()
fig.savefig(buffer, format="png", bbox_inches="tight")
plt.close(fig)
display(Image(data=buffer.getvalue()))
```
