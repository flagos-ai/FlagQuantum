## Dense preparation by rotations

The target starts in $|0\rangle$. For a normalized power-of-two coefficient vector, prepare magnitudes one target bit at a time, then encode the coefficient phases. The rotation tree quantizes the angles into words of the selected phase width. A programmable phase-gradient lookup loads each word, adds it into the gradient, and erases the loaded word. The gradient eigenstate returns unchanged.

The operation acts when all `prepare_control` bits are one. The target and gradient words are big-endian through ports. The phase flag and lookup work are temporary. Select `uncompute=True` to apply the adjoint.

- [Low, Kliuchnikov and Schaeffer, Trading T-gates for dirty qubits in state preparation and unitary synthesis](https://arxiv.org/abs/1812.00954)

### Example Instances

```python
op = RotationPreparationExample()
op.coefficients
```

```python
length = Symbol("L")
phase = Symbol(r"	ext{phase_bitsize}")
shape_signature = rotation_signature(length, phase)
source_rotation_preparation_resource_formula(length, phase)
```

```python
gradient_bits = Symbol(r"b_	ext{grad}")
gradient_signature = rotation_signature(len(op.coefficients), gradient_bits)
source_rotation_preparation_resource_formula(len(op.coefficients), gradient_bits)
```
